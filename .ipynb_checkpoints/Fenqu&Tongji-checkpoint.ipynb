```python
# -*- coding: utf-8 -*-
import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd
import os
from matplotlib.patches import Patch

# 读取站点数据并去重
def read_and_process_stations(csv_path):
    df = pd.read_csv(csv_path)
    required_columns = ['station', 'longitude', 'latitude']
    for col in required_columns:
        if col not in df.columns:
            raise ValueError(f"CSV文件中缺少必要的列: {col}")
    df_unique = df.drop_duplicates(subset=['station'], keep='first')
    print(f"原始数据站点数: {len(df)}, 去重后: {len(df_unique)}")
    return df_unique

# 创建站点GeoDataFrame
def create_stations_gdf(df):
    geometry = gpd.points_from_xy(df.longitude, df.latitude)
    gdf_stations = gpd.GeoDataFrame(df, geometry=geometry, crs="EPSG:4326")
    return gdf_stations

# 设置路径
csv_path = r"D:\TUB\greenlandProject\NEWdataverse_files\allsites_albedoYM_lonlai.csv"
folder_path = r"C:\greenland\Greenland_shp\流域冰盖\new"

out_polygons = r"C:\greenland\Greenland_shp\合并分区.shp"
out_points   = r"C:\greenland\Greenland_shp\站点.shp"
out_gpkg     = r"C:\greenland\Greenland_shp\总合并结果.gpkg"  # 推荐：支持混合几何

# 1. 站点
df_stations = read_and_process_stations(csv_path)
gdf_stations = create_stations_gdf(df_stations)

# 2. 合并文件夹下的分区shp
shp_files = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.endswith('.shp')]
if not shp_files:
    raise FileNotFoundError("文件夹中没有找到任何shapefile文件。")

gdfs = []
for fp in shp_files:
    gdf = gpd.read_file(fp)
    if gdf.crs != gdf_stations.crs:
        gdf = gdf.to_crs(gdf_stations.crs)
    gdf["source"] = os.path.splitext(os.path.basename(fp))[0]
    gdfs.append(gdf)

gdf_polygons = gpd.GeoDataFrame(pd.concat(gdfs, ignore_index=True), crs=gdf_stations.crs)
print(f"✅ 合并后的分区要素数: {len(gdf_polygons)}")

# 3. 保存为独立文件
gdf_polygons.to_file(out_polygons, encoding="utf-8")
print(f"✅ 分区保存到: {out_polygons}")

gdf_stations.to_file(out_points, encoding="utf-8")
print(f"✅ 站点保存到: {out_points}")

# 4. 保存为 GeoPackage (可同时存点和面)
gdf_polygons.to_file(out_gpkg, layer="polygons", driver="GPKG")
gdf_stations.to_file(out_gpkg, layer="stations", driver="GPKG")
print(f"✅ 混合结果保存到: {out_gpkg} (包含两层: polygons + stations)")

# 5. 可视化
fig, ax = plt.subplots(1, 1, figsize=(15, 12))
colors = plt.cm.tab20.colors
legend_elements = []
for i, (name, g) in enumerate(gdf_polygons.groupby("source")):
    color = colors[i % len(colors)]
    g.plot(ax=ax, facecolor=color, edgecolor="black", alpha=0.6)
    legend_elements.append(Patch(facecolor=color, edgecolor="black", alpha=0.6, label=name))

gdf_stations.plot(ax=ax, color="red", marker="o", markersize=50,
                  edgecolor="black", linewidth=1, label="监测站点")
legend_elements.append(Patch(facecolor='red', edgecolor='black',
                             label=f'监测站点 ({len(gdf_stations)}个)'))

ax.legend(handles=legend_elements, loc="upper right", fontsize=10)
ax.set_title("格陵兰冰盖分区 + 站点", fontsize=16, fontproperties="SimHei")
ax.set_xlabel("经度", fontproperties="SimHei")
ax.set_ylabel("纬度", fontproperties="SimHei")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# -*- coding: utf-8 -*-
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
from shapely.geometry import Point

# === 路径设置 ===
csv_path = r"C:\greenland\1995bu\csv_grid_mk\albedo_MK_AMJJAS_1950-2024_grid.csv"
shp_path = r"C:\greenland\Greenland_shp\合并分区.shp"   
stations_path = r"C:\greenland\Greenland_shp\站点.shp"   # 你保存过的站点数据
out_path = r"C:\greenland\Greenland_shp\趋势结果.gpkg"

# === 1. 读入格点趋势数据 ===
df = pd.read_csv(csv_path)
df = df.dropna(subset=["slope_decade"])
geometry = [Point(xy) for xy in zip(df["lon"], df["lat"])]
gdf_points = gpd.GeoDataFrame(df, geometry=geometry, crs="EPSG:4326")

# === 2. 读入分区图层 ===
gdf_poly = gpd.read_file(shp_path)
if gdf_poly.crs != gdf_points.crs:
    gdf_poly = gdf_poly.to_crs(gdf_points.crs)

# === 3. 空间连接（格点 -> 分区） ===
gdf_joined = gpd.sjoin(gdf_points, gdf_poly, how="left", predicate="within")

# === 4. 读入站点 ===
gdf_stations = gpd.read_file(stations_path)
if gdf_stations.crs != gdf_points.crs:
    gdf_stations = gdf_stations.to_crs(gdf_points.crs)

# === 5. 可视化 ===
fig, ax = plt.subplots(1, 1, figsize=(12, 10))

# (a) 分区边界
gdf_poly.boundary.plot(ax=ax, color="black", linewidth=0.3)   # 线条更细

# (b) 趋势点
gdf_joined.plot(
    ax=ax,
    column="slope_decade",
    cmap="bwr",
    vmin=-0.01, vmax=0.01,
    markersize=5,
    legend=True,
    legend_kwds={"label": "Albedo趋势 (每十年变化)"}
)

# (c) 站点 - 醒目的圆形标注
gdf_stations.plot(
    ax=ax, color="gold", marker="o", markersize=90,
    edgecolor="darkred", linewidth=0.8, label="监测站点"
)

ax.set_title("格陵兰 Albedo 趋势 + 监测站点", fontsize=14, fontweight="bold")
ax.set_xlabel("经度")
ax.set_ylabel("纬度")
ax.legend(loc="upper right", fontsize=9)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# === 6. 保存结果 ===
gdf_poly.to_file(out_path, layer="polygons", driver="GPKG")
gdf_joined.to_file(out_path, layer="albedo_trend_points", driver="GPKG")
gdf_stations.to_file(out_path, layer="stations", driver="GPKG")
print(f"✅ 已保存趋势结果到 {out_path} (含 polygons + trend_points + stations 三个图层)")
```

```python
# -*- coding: utf-8 -*-
import xarray as xr
import geopandas as gpd
import matplotlib.pyplot as plt
import rioxarray  # 确保已安装 pip install rioxarray geopandas shapely
import pandas as pd

# 文件路径
nc_path = r"C:\greenland\1995bu\Greenland_albedo1950-2024.nc"
shp_dir = r"C:\greenland\Greenland_shp\流域冰盖\new"   # 假设这里有 xiaorong.shp, dong.shp, xi.shp

# === 1. 读取 NetCDF 数据 ===
ds = xr.open_dataset(nc_path)
albedo = ds["albedo"]

# 只取夏季 (6,7,8 月)
albedo_summer = albedo.where(albedo["time.month"].isin([6, 7, 8]), drop=True)

# === 2. 读取分区 shapefile ===
regions = {
    "xiaorong": gpd.read_file(f"{shp_dir}/xiaorong.shp"),
    "dong": gpd.read_file(f"{shp_dir}/dong.shp"),
    "xi": gpd.read_file(f"{shp_dir}/xi.shp"),
}

# 结果保存
results = {}

# === 3. 每个分区计算夏季均值 ===
for name, gdf in regions.items():
    # CRS 确保一致
    if gdf.crs != "EPSG:4326":
        gdf = gdf.to_crs("EPSG:4326")
    geom = gdf.geometry.unary_union

    # rioxarray 裁剪
    albedo_clip = albedo_summer.rio.write_crs("EPSG:4326").rio.clip([geom], all_touched=True)

    # 按年份聚合
    df_mean = (
        albedo_clip.groupby("time.year").mean(dim=["time", "lat", "lon"]).to_dataframe().reset_index()
    )
    df_mean = df_mean.rename(columns={"albedo": name})
    results[name] = df_mean.set_index("year")[name]

# === 4. 合并所有分区结果 ===
df_all = pd.concat(results.values(), axis=1)
print(df_all.head())

# === 5. 画折线图 ===
plt.figure(figsize=(12, 6))
for name in results.keys():
    plt.plot(df_all.index, df_all[name], label=name, linewidth=1.5)

plt.title("1950–2024 格陵兰各分区夏季 (JJA) Albedo 均值", fontsize=14, fontweight="bold")
plt.xlabel("年份")
plt.ylabel("Albedo 均值")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# -*- coding: utf-8 -*-
import xarray as xr
import geopandas as gpd
import matplotlib.pyplot as plt
import rioxarray
import pandas as pd
import numpy as np
from scipy import stats
from sklearn.linear_model import LinearRegression
import warnings
warnings.filterwarnings('ignore')

# 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# 文件路径
nc_path = r"C:\greenland\1995bu\Greenland_albedo1950-2024.nc"
shp_dir = r"C:\greenland\Greenland_shp\流域冰盖\new"

print("="*60)
print("格陵兰流域 Albedo 趋势分析与异常检测")
print("="*60)

# === STEP 1: 读取 NetCDF 数据 ===
print("\nSTEP 1: 读取数据...")
ds = xr.open_dataset(nc_path)
albedo = ds["albedo"]

# 只取夏季 (6,7,8 月)
albedo_summer = albedo.where(albedo["time.month"].isin([6, 7, 8]), drop=True)
print(f"数据时间范围: {albedo_summer.time.dt.year.min().values} - {albedo_summer.time.dt.year.max().values}")

# === STEP 2: 读取分区 shapefile ===
print("\nSTEP 2: 读取流域边界...")
regions = {
    "xiaorong": gpd.read_file(f"{shp_dir}/xiaorong.shp"),
    "dong": gpd.read_file(f"{shp_dir}/dong.shp"),
    "xi": gpd.read_file(f"{shp_dir}/xi.shp"),
}

# 结果保存
results = {}

# === STEP 3: 每个分区计算夏季均值 ===
print("\nSTEP 3: 计算各流域夏季 Albedo 均值...")
for name, gdf in regions.items():
    print(f"  处理 {name} 流域...")
    
    # CRS 确保一致
    if gdf.crs != "EPSG:4326":
        gdf = gdf.to_crs("EPSG:4326")
    geom = gdf.geometry.unary_union
    
    # rioxarray 裁剪
    albedo_clip = albedo_summer.rio.write_crs("EPSG:4326").rio.clip([geom], all_touched=True)
    
    # 按年份聚合
    df_mean = (
        albedo_clip.groupby("time.year").mean(dim=["time", "lat", "lon"]).to_dataframe().reset_index()
    )
    df_mean = df_mean.rename(columns={"albedo": name})
    results[name] = df_mean.set_index("year")[name]

# 合并所有分区结果
df_all = pd.concat(results.values(), axis=1)
print(f"数据处理完成，共 {len(df_all)} 年数据")

# === STEP 4: 趋势分析函数 ===
def calculate_trend(series, years):
    """计算线性趋势"""
    X = years.values.reshape(-1, 1)
    y = series.values
    
    # 去除 NaN 值
    mask = ~np.isnan(y)
    if mask.sum() < 3:  # 至少需要3个点
        return np.nan, np.nan, np.nan, np.nan
    
    X_clean = X[mask]
    y_clean = y[mask]
    
    model = LinearRegression()
    model.fit(X_clean, y_clean)
    
    slope = model.coef_[0]
    intercept = model.intercept_
    r_squared = model.score(X_clean, y_clean)
    
    # 计算 p 值
    slope_scipy, intercept_scipy, r_value, p_value, std_err = stats.linregress(
        X_clean.flatten(), y_clean
    )
    
    return slope, intercept, r_squared, p_value

def detect_anomalies(series, threshold=2.0):
    """检测异常值（基于 Z-score）"""
    z_scores = np.abs(stats.zscore(series.dropna()))
    return series.index[z_scores > threshold].tolist()

def calculate_variability_change(series, split_year=2005):
    """计算变异性变化（前后两段的标准差比较）"""
    early = series[series.index < split_year]
    recent = series[series.index >= split_year]
    
    if len(early) < 3 or len(recent) < 3:
        return np.nan, np.nan, np.nan
    
    early_std = early.std()
    recent_std = recent.std()
    variability_ratio = recent_std / early_std if early_std > 0 else np.nan
    
    # F检验比较方差
    f_stat = recent.var() / early.var() if early.var() > 0 else np.nan
    f_p_value = 1 - stats.f.cdf(f_stat, len(recent)-1, len(early)-1) if not np.isnan(f_stat) else np.nan
    
    return early_std, recent_std, variability_ratio, f_p_value

# === STEP 5: 趋势分析 ===
print("\nSTEP 5: 趋势分析...")

# 定义时间段
years_all = df_all.index
years_recent = df_all.index[df_all.index >= 2005]  # 近20年

stats_table = []

for region in df_all.columns:
    print(f"\n--- {region} 流域分析 ---")
    
    series = df_all[region]
    
    # 1. 总体趋势 (全时间段)
    slope_all, _, r2_all, p_all = calculate_trend(series, years_all)
    
    # 2. 近20年趋势
    series_recent = series[series.index >= 2005]
    slope_recent, _, r2_recent, p_recent = calculate_trend(series_recent, years_recent)
    
    # 3. 分段趋势 (1950-1990, 1990-2010, 2010-2024)
    periods = {
        "1950-1990": (1950, 1990),
        "1990-2010": (1990, 2010), 
        "2010-2024": (2010, 2024)
    }
    
    period_trends = {}
    for period_name, (start, end) in periods.items():
        period_data = series[(series.index >= start) & (series.index <= end)]
        if len(period_data) >= 3:
            slope, _, r2, p = calculate_trend(period_data, period_data.index)
            period_trends[period_name] = {
                'slope': slope, 'r2': r2, 'p_value': p, 'years': len(period_data)
            }
    
    # 4. 变异性分析
    early_std, recent_std, var_ratio, f_p = calculate_variability_change(series, 2005)
    
    # 5. 异常值检测
    anomalies = detect_anomalies(series, threshold=2.0)
    
    # 6. 计算加速度（趋势变化）
    # 比较前后两个时期的趋势斜率
    early_period = series[series.index < 2005]
    recent_period = series[series.index >= 2005]
    
    slope_early, _, _, _ = calculate_trend(early_period, early_period.index)
    slope_recent_calc, _, _, _ = calculate_trend(recent_period, recent_period.index)
    trend_acceleration = slope_recent_calc - slope_early if not np.isnan(slope_early) and not np.isnan(slope_recent_calc) else np.nan
    
    # 保存统计结果
    stats_row = {
        '流域': region,
        '总体趋势斜率': slope_all,
        '总体R²': r2_all,
        '总体P值': p_all,
        '近20年趋势斜率': slope_recent,
        '近20年R²': r2_recent,
        '近20年P值': p_recent,
        '前期标准差(2005前)': early_std,
        '近期标准差(2005后)': recent_std,
        '变异性比值': var_ratio,
        '方差变化P值': f_p,
        '趋势加速度': trend_acceleration,
        '异常年份数量': len(anomalies),
        '异常年份': ','.join(map(str, anomalies)) if anomalies else '无'
    }
    
    # 添加分段趋势
    for period_name, trend_data in period_trends.items():
        stats_row[f'{period_name}_斜率'] = trend_data['slope']
        stats_row[f'{period_name}_R²'] = trend_data['r2']
        stats_row[f'{period_name}_P值'] = trend_data['p_value']
    
    stats_table.append(stats_row)

# === STEP 6: 创建统计表格 ===
print("\nSTEP 6: 生成统计表格...")
df_stats = pd.DataFrame(stats_table)

# 格式化数值显示
pd.set_option('display.float_format', '{:.6f}'.format)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

print("\n" + "="*100)
print("统计分析结果表格")
print("="*100)
print(df_stats.to_string(index=False))

# === STEP 7: 异常年份详细分析 ===
print("\n" + "="*60)
print("异常年份详细分析")
print("="*60)

for region in df_all.columns:
    series = df_all[region]
    anomalies = detect_anomalies(series, threshold=2.0)
    
    if anomalies:
        print(f"\n{region} 流域异常年份:")
        for year in anomalies:
            z_score = np.abs(stats.zscore(series.dropna()))[series.index.get_loc(year)]
            albedo_val = series[year]
            mean_val = series.mean()
            deviation = albedo_val - mean_val
            print(f"  {year}: Albedo={albedo_val:.4f} (均值{deviation:+.4f}, Z-score={z_score:.2f})")
    else:
        print(f"\n{region} 流域: 无显著异常年份")

# === STEP 8: 变异性趋势检验 ===
print("\n" + "="*60)
print("变异性变化分析")
print("="*60)

for region in df_all.columns:
    series = df_all[region]
    early_std, recent_std, var_ratio, f_p = calculate_variability_change(series, 2005)
    
    print(f"\n{region} 流域:")
    print(f"  前期变异性 (2005前): {early_std:.6f}")
    print(f"  近期变异性 (2005后): {recent_std:.6f}")
    print(f"  变异性比值: {var_ratio:.3f}")
    
    if f_p < 0.05:
        change_type = "显著增加" if var_ratio > 1 else "显著减少"
        print(f"  变异性{change_type} (P={f_p:.4f})")
    else:
        print(f"  变异性无显著变化 (P={f_p:.4f})")

# === STEP 9: 综合可视化 ===
print("\nSTEP 9: 生成综合分析图...")

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('格陵兰流域 Albedo 综合分析 (1950-2024)', fontsize=16, fontweight='bold')

# 子图1: 原始时间序列 + 趋势线
ax1 = axes[0, 0]
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']

for i, region in enumerate(df_all.columns):
    series = df_all[region]
    
    # 绘制原始数据
    ax1.plot(series.index, series.values, 'o-', color=colors[i], 
             label=f'{region}', linewidth=1.5, markersize=4, alpha=0.7)
    
    # 添加总体趋势线
    slope, intercept, r2, p = calculate_trend(series, series.index)
    if not np.isnan(slope):
        trend_line = slope * series.index + intercept
        ax1.plot(series.index, trend_line, '--', color=colors[i], 
                linewidth=2, alpha=0.8)

ax1.set_title('夏季 Albedo 时间序列及趋势线')
ax1.set_xlabel('年份')
ax1.set_ylabel('Albedo 均值')
ax1.legend()
ax1.grid(alpha=0.3)

# 子图2: 滑动标准差 (变异性)
ax2 = axes[0, 1]
window = 10  # 10年滑动窗口

for i, region in enumerate(df_all.columns):
    series = df_all[region]
    rolling_std = series.rolling(window=window, center=True).std()
    ax2.plot(rolling_std.index, rolling_std.values, 'o-', color=colors[i], 
             label=f'{region}', linewidth=1.5, markersize=3)

ax2.axvline(x=2005, color='red', linestyle='--', alpha=0.7, label='2005分界线')
ax2.set_title(f'{window}年滑动标准差 (变异性趋势)')
ax2.set_xlabel('年份')
ax2.set_ylabel('滑动标准差')
ax2.legend()
ax2.grid(alpha=0.3)

# 子图3: 分段趋势比较
ax3 = axes[1, 0]
periods = ['1950-1990', '1990-2010', '2010-2024']
x_pos = np.arange(len(periods))
width = 0.25

for i, region in enumerate(df_all.columns):
    slopes = []
    for period in periods:
        start, end = map(int, period.split('-'))
        period_data = df_all[region][(df_all.index >= start) & (df_all.index <= end)]
        if len(period_data) >= 3:
            slope, _, _, _ = calculate_trend(period_data, period_data.index)
            slopes.append(slope * 10)  # 转换为每10年变化
        else:
            slopes.append(0)
    
    ax3.bar(x_pos + i*width, slopes, width, label=region, color=colors[i], alpha=0.7)

ax3.set_title('分段趋势比较 (每10年变化率)')
ax3.set_xlabel('时间段')
ax3.set_ylabel('Albedo 变化率 (/10年)')
ax3.set_xticks(x_pos + width)
ax3.set_xticklabels(periods)
ax3.legend()
ax3.grid(axis='y', alpha=0.3)
ax3.axhline(y=0, color='black', linestyle='-', linewidth=0.5)

# 子图4: 异常年份标识
ax4 = axes[1, 1]

for i, region in enumerate(df_all.columns):
    series = df_all[region]
    
    # 绘制标准化数据
    normalized = (series - series.mean()) / series.std()
    ax4.plot(series.index, normalized, 'o-', color=colors[i], 
             label=f'{region}', linewidth=1.5, markersize=4, alpha=0.7)
    
    # 标记异常值
    anomalies = detect_anomalies(series, threshold=2.0)
    for year in anomalies:
        if year in normalized.index:
            ax4.plot(year, normalized[year], 'x', color='red', markersize=10, markeredgewidth=3)

# 添加阈值线
ax4.axhline(y=2, color='red', linestyle='--', alpha=0.5, label='异常阈值 (±2σ)')
ax4.axhline(y=-2, color='red', linestyle='--', alpha=0.5)
ax4.axhline(y=0, color='black', linestyle='-', linewidth=0.5, alpha=0.5)

ax4.set_title('标准化 Albedo 与异常检测')
ax4.set_xlabel('年份')
ax4.set_ylabel('标准化 Albedo (Z-score)')
ax4.legend()
ax4.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# === STEP 10: 生成综合统计报告 ===
print("\n" + "="*80)
print("综合统计分析报告")
print("="*80)

# 创建详细统计表
detailed_stats = []

for region in df_all.columns:
    series = df_all[region]
    
    # 基础统计
    mean_all = series.mean()
    std_all = series.std()
    min_val = series.min()
    max_val = series.max()
    
    # 趋势分析
    slope_all, _, r2_all, p_all = calculate_trend(series, series.index)
    slope_recent, _, r2_recent, p_recent = calculate_trend(
        series[series.index >= 2005], series[series.index >= 2005].index
    )
    
    # 变异性分析
    early_std, recent_std, var_ratio, f_p = calculate_variability_change(series, 2005)
    
    # 异常检测
    anomalies = detect_anomalies(series, threshold=2.0)
    
    # 趋势显著性判断
    trend_significance_all = "显著" if p_all < 0.05 else "不显著"
    trend_significance_recent = "显著" if p_recent < 0.05 else "不显著"
    
    # 变异性变化判断
    if f_p < 0.05:
        variability_change = "显著增加" if var_ratio > 1 else "显著减少"
    else:
        variability_change = "无显著变化"
    
    row = {
        '流域': region,
        '时间序列均值': f"{mean_all:.6f}",
        '时间序列标准差': f"{std_all:.6f}",
        '最小值(年份)': f"{min_val:.6f}({series.idxmin()})",
        '最大值(年份)': f"{max_val:.6f}({series.idxmax()})",
        '总体趋势(每年)': f"{slope_all:.8f}",
        '总体趋势(每10年)': f"{slope_all*10:.6f}",
        '总体趋势R²': f"{r2_all:.4f}",
        '总体趋势显著性': f"{trend_significance_all}(P={p_all:.4f})",
        '近20年趋势(每年)': f"{slope_recent:.8f}",
        '近20年趋势(每10年)': f"{slope_recent*10:.6f}",
        '近20年趋势R²': f"{r2_recent:.4f}",
        '近20年趋势显著性': f"{trend_significance_recent}(P={p_recent:.4f})",
        '前期变异性(2005前)': f"{early_std:.6f}",
        '近期变异性(2005后)': f"{recent_std:.6f}",
        '变异性比值': f"{var_ratio:.3f}",
        '变异性变化': f"{variability_change}(P={f_p:.4f})",
        '异常年份数': len(anomalies),
        '异常年份': ','.join(map(str, anomalies)) if anomalies else '无',
        '数据完整性': f"{(~series.isna()).sum()}/{len(series)}年"
    }
    
    detailed_stats.append(row)

# 转换为 DataFrame 并显示
df_detailed = pd.DataFrame(detailed_stats)

print("\n详细统计表格:")
print("-" * 120)
for col in df_detailed.columns:
    print(f"{col}:")
    for _, row in df_detailed.iterrows():
        print(f"  {row['流域']:>8}: {row[col]}")
    print()

# === STEP 11: 风险评估与预警 ===
print("="*80)
print("风险评估与预警")
print("="*80)

for region in df_all.columns:
    series = df_all[region]
    
    # 计算风险指标
    risk_factors = []
    
    # 1. 趋势风险
    slope_recent, _, _, p_recent = calculate_trend(
        series[series.index >= 2005], series[series.index >= 2005].index
    )
    if p_recent < 0.05 and slope_recent < -0.001:  # 显著下降趋势
        risk_factors.append(f"显著下降趋势({slope_recent*10:.4f}/10年)")
    
    # 2. 变异性风险
    _, _, var_ratio, f_p = calculate_variability_change(series, 2005)
    if f_p < 0.05 and var_ratio > 1.5:  # 变异性显著增加
        risk_factors.append(f"变异性显著增加({var_ratio:.2f}倍)")
    
    # 3. 异常频率风险
    recent_anomalies = [year for year in detect_anomalies(series) if year >= 2010]
    if len(recent_anomalies) >= 3:  # 近15年有3次以上异常
        risk_factors.append(f"近期异常频繁({len(recent_anomalies)}次)")
    
    # 4. 极值风险
    recent_series = series[series.index >= 2010]
    if recent_series.min() == series.min():  # 历史最低值出现在近期
        risk_factors.append("历史最低值出现在近期")
    
    print(f"\n{region} 流域风险评估:")
    if risk_factors:
        print("  ⚠️  发现以下风险因子:")
        for factor in risk_factors:
            print(f"     • {factor}")
        
        # 综合风险等级
        risk_level = len(risk_factors)
        if risk_level >= 3:
            print("  🔴 综合风险等级: 高风险")
        elif risk_level == 2:
            print("  🟡 综合风险等级: 中等风险")
        else:
            print("  🟢 综合风险等级: 低风险")
    else:
        print("  ✅ 未发现显著风险因子")
        print("  🟢 综合风险等级: 低风险")

# === STEP 12: 保存结果 ===
print(f"\n{'='*60}")
print("分析完成")
print("="*60)

# 保存统计表格到 CSV
output_path = r"C:\greenland\albedo_analysis_results.csv"
try:
    df_detailed.to_csv(output_path, index=False, encoding='utf-8-sig')
    print(f"✅ 详细统计结果已保存至: {output_path}")
except:
    print("📁 无法保存文件，请检查路径权限")

print(f"✅ 分析完成! 共处理 {len(df_all.columns)} 个流域，{len(df_all)} 年数据")
print("📊 图表已显示，请查看统计表格进行进一步分析")
```

```python
# -*- coding: utf-8 -*-
import xarray as xr
import rioxarray  # pip install rioxarray
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt

# ========= 路径 =========
nc_path   = r"C:\greenland\1995bu\Greenland_albedo1950-2024.nc"
bands_shp = r"C:\greenland\Greenland_shp\高程带_500m_五区域.shp"
out_csv   = r"C:\greenland\Greenland_shp\JJA_albedo_by_5bands.csv"

# ========= 1) 读取高程带矢量 =========
bands = gpd.read_file(bands_shp)
# 统一到 WGS84，后续给 NetCDF 裁剪
if bands.crs is None or bands.crs.to_string().upper() != "EPSG:4326":
    bands = bands.to_crs("EPSG:4326")

# 规范顺序（若缺带则自动跳过）
band_order = ["<1500m", "1500-2000m", "2000-2500m", "2500-3000m", ">=3000m"]
bands["band"] = bands["band"].astype(str)
bands = bands[bands["band"].isin(band_order)]
bands["order"] = bands["band"].map({b:i for i,b in enumerate(band_order)})
bands = bands.sort_values("order")

print("将计算的高程带顺序：", bands["band"].tolist())

# ========= 2) 读取 NetCDF 并选择 JJA =========
ds = xr.open_dataset(nc_path)
albedo = ds["albedo"]
# 只取 6-8 月
albedo_jja = albedo.where(albedo["time"].dt.month.isin([6,7,8]), drop=True)
# 写入 CRS（rioxarray 需要）
albedo_jja = albedo_jja.rio.write_crs("EPSG:4326", inplace=False)

# ========= 3) 对每个高程带裁剪并按年求均值 =========
series_dict = {}
for _, row in bands.iterrows():
    band_name = row["band"]
    geom = row.geometry

    # 裁剪到该高程带
    clipped = albedo_jja.rio.clip([geom], all_touched=True)

    # 对每年：先对 JJA 时间平均，再做空间均值
    # 也可以先空间均值再时间均值，结果一致（线性算子）
    df_band = (
        clipped.groupby("time.year")
               .mean(dim=["time", "lat", "lon"])
               .to_dataframe()
               .reset_index()
               .rename(columns={"albedo": band_name})
               .set_index("year")[band_name]
    )
    series_dict[band_name] = df_band

# 合并为一个 DataFrame
df_all = pd.concat([series_dict[b] for b in band_order if b in series_dict], axis=1)
df_all.index.name = "year"
print("预览：")
print(df_all.head())

# 保存结果以便重启后直接读取
df_all.to_csv(out_csv, float_format="%.6f", encoding="utf-8-sig")
print(f"✅ 已保存 JJA albedo 均值到：{out_csv}")

# ========= 4) 绘制五条折线 =========
plt.figure(figsize=(12, 6))
for b in [c for c in band_order if c in df_all.columns]:
    plt.plot(df_all.index, df_all[b], label=b, linewidth=1.6)

plt.title("1950–2024 各高程带夏季 (JJA) Albedo 年均值", fontsize=14, fontweight="bold")
plt.xlabel("年份")
plt.ylabel("Albedo 均值")
plt.grid(alpha=0.3)
plt.legend(title="高程带", ncol=2, fontsize=9)
plt.tight_layout()
plt.show()
```

```python
# -*- coding: utf-8 -*-
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
from sklearn.linear_model import LinearRegression
import warnings
warnings.filterwarnings('ignore')

# 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# 文件路径
csv_path = r"C:\greenland\Greenland_shp\JJA_albedo_by_5bands.csv"

print("="*60)
print("格陵兰不同海拔带 Albedo 趋势分析与异常检测")
print("="*60)

# === STEP 1: 读取 CSV 数据 ===
print("\nSTEP 1: 读取数据...")
try:
    df_all = pd.read_csv(csv_path)
    df_all = df_all.set_index('year')
    print("数据读取成功！")
    print(f"数据时间范围: {df_all.index.min()} - {df_all.index.max()}")
    print(f"分析的海拔带: {', '.join(df_all.columns)}")
except FileNotFoundError:
    print(f"错误: 找不到文件 {csv_path}。请检查文件路径。")
    exit()

# === STEP 2: 趋势分析函数 (保持不变) ===
def calculate_trend(series, years):
    """计算线性趋势"""
    X = years.values.reshape(-1, 1)
    y = series.values
    
    # 去除 NaN 值
    mask = ~np.isnan(y)
    if mask.sum() < 3:  # 至少需要3个点
        return np.nan, np.nan, np.nan, np.nan
    
    X_clean = X[mask]
    y_clean = y[mask]
    
    model = LinearRegression()
    model.fit(X_clean, y_clean)
    
    slope = model.coef_[0]
    intercept = model.intercept_
    r_squared = model.score(X_clean, y_clean)
    
    # 计算 p 值
    slope_scipy, intercept_scipy, r_value, p_value, std_err = stats.linregress(
        X_clean.flatten(), y_clean
    )
    
    return slope, intercept, r_squared, p_value

def detect_anomalies(series, threshold=2.0):
    """检测异常值（基于 Z-score）"""
    z_scores = np.abs(stats.zscore(series.dropna()))
    # 使用 .index.to_series() 来安全地索引
    return series.dropna().index[z_scores > threshold].tolist()

def calculate_variability_change(series, split_year=2005):
    """计算变异性变化（前后两段的标准差比较）"""
    early = series[series.index < split_year]
    recent = series[series.index >= split_year]
    
    if len(early) < 3 or len(recent) < 3:
        return np.nan, np.nan, np.nan, np.nan
    
    early_std = early.std()
    recent_std = recent.std()
    variability_ratio = recent_std / early_std if early_std > 0 else np.nan
    
    # F检验比较方差
    f_stat = recent.var() / early.var() if early.var() > 0 else np.nan
    f_p_value = 1 - stats.f.cdf(f_stat, len(recent)-1, len(early)-1) if not np.isnan(f_stat) else np.nan
    
    return early_std, recent_std, variability_ratio, f_p_value

# === STEP 3: 趋势分析 ===
print("\nSTEP 3: 趋势分析...")

# 定义时间段
years_all = df_all.index
years_recent = df_all.index[df_all.index >= 2005]  # 近20年

stats_table = []

for band in df_all.columns:
    print(f"\n--- {band} 海拔带分析 ---")
    
    series = df_all[band]
    
    # 1. 总体趋势 (全时间段)
    slope_all, _, r2_all, p_all = calculate_trend(series, years_all)
    
    # 2. 近20年趋势
    series_recent = series[series.index >= 2005]
    slope_recent, _, r2_recent, p_recent = calculate_trend(series_recent, years_recent)
    
    # 3. 分段趋势 (1950-1990, 1990-2010, 2010-2024)
    periods = {
        "1950-1990": (1950, 1990),
        "1990-2010": (1990, 2010), 
        "2010-2024": (2010, 2024)
    }
    
    period_trends = {}
    for period_name, (start, end) in periods.items():
        period_data = series[(series.index >= start) & (series.index <= end)]
        if len(period_data) >= 3:
            slope, _, r2, p = calculate_trend(period_data, period_data.index)
            period_trends[period_name] = {
                'slope': slope, 'r2': r2, 'p_value': p, 'years': len(period_data)
            }
    
    # 4. 变异性分析
    early_std, recent_std, var_ratio, f_p = calculate_variability_change(series, 2005)
    
    # 5. 异常值检测
    anomalies = detect_anomalies(series, threshold=2.0)
    
    # 6. 计算加速度（趋势变化）
    early_period = series[series.index < 2005]
    recent_period = series[series.index >= 2005]
    
    slope_early, _, _, _ = calculate_trend(early_period, early_period.index)
    slope_recent_calc, _, _, _ = calculate_trend(recent_period, recent_period.index)
    trend_acceleration = slope_recent_calc - slope_early if not np.isnan(slope_early) and not np.isnan(slope_recent_calc) else np.nan
    
    # 保存统计结果
    stats_row = {
        '海拔带': band,
        '总体趋势斜率': slope_all,
        '总体R²': r2_all,
        '总体P值': p_all,
        '近20年趋势斜率': slope_recent,
        '近20年R²': r2_recent,
        '近20年P值': p_recent,
        '前期标准差(2005前)': early_std,
        '近期标准差(2005后)': recent_std,
        '变异性比值': var_ratio,
        '方差变化P值': f_p,
        '趋势加速度': trend_acceleration,
        '异常年份数量': len(anomalies),
        '异常年份': ','.join(map(str, anomalies)) if anomalies else '无'
    }
    
    # 添加分段趋势
    for period_name, trend_data in period_trends.items():
        stats_row[f'{period_name}_斜率'] = trend_data['slope']
        stats_row[f'{period_name}_R²'] = trend_data['r2']
        stats_row[f'{period_name}_P值'] = trend_data['p_value']
    
    stats_table.append(stats_row)

# === STEP 4: 创建统计表格 ===
print("\nSTEP 4: 生成统计表格...")
df_stats = pd.DataFrame(stats_table)

# 格式化数值显示
pd.set_option('display.float_format', '{:.6f}'.format)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

print("\n" + "="*100)
print("统计分析结果表格")
print("="*100)
print(df_stats.to_string(index=False))

# === STEP 5: 异常年份详细分析 ===
print("\n" + "="*60)
print("异常年份详细分析")
print("="*60)

for band in df_all.columns:
    series = df_all[band]
    anomalies = detect_anomalies(series, threshold=2.0)
    
    if anomalies:
        print(f"\n{band} 海拔带异常年份:")
        # 使用 .dropna() 确保 z-score 和 series 的索引对齐
        series_clean = series.dropna()
        z_scores_clean = np.abs(stats.zscore(series_clean))
        z_scores_series = pd.Series(z_scores_clean, index=series_clean.index)

        for year in anomalies:
            albedo_val = series[year]
            mean_val = series.mean()
            deviation = albedo_val - mean_val
            z_score = z_scores_series.get(year, np.nan)
            print(f"  {year}: Albedo={albedo_val:.4f} (均值{deviation:+.4f}, Z-score={z_score:.2f})")
    else:
        print(f"\n{band} 海拔带: 无显著异常年份")

# === STEP 6: 变异性趋势检验 ===
print("\n" + "="*60)
print("变异性变化分析")
print("="*60)

for band in df_all.columns:
    series = df_all[band]
    early_std, recent_std, var_ratio, f_p = calculate_variability_change(series, 2005)
    
    print(f"\n{band} 海拔带:")
    print(f"  前期变异性 (2005前): {early_std:.6f}")
    print(f"  近期变异性 (2005后): {recent_std:.6f}")
    print(f"  变异性比值: {var_ratio:.3f}")
    
    if not np.isnan(f_p) and f_p < 0.05:
        change_type = "显著增加" if var_ratio > 1 else "显著减少"
        print(f"  变异性{change_type} (P={f_p:.4f})")
    else:
        print(f"  变异性无显著变化 (P={f_p:.4f})")

# === STEP 7: 综合可视化 ===
print("\nSTEP 7: 生成综合分析图...")

fig, axes = plt.subplots(2, 2, figsize=(18, 14))
fig.suptitle('格陵兰不同海拔带 Albedo 综合分析 (1950-2024)', fontsize=18, fontweight='bold')

# 扩展颜色列表以适应5个海拔带
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']

# 子图1: 原始时间序列 + 趋势线
ax1 = axes[0, 0]
for i, band in enumerate(df_all.columns):
    series = df_all[band]
    
    # 绘制原始数据
    ax1.plot(series.index, series.values, 'o-', color=colors[i], 
             label=f'{band}', linewidth=1.5, markersize=4, alpha=0.7)
    
    # 添加总体趋势线
    slope, intercept, r2, p = calculate_trend(series, series.index)
    if not np.isnan(slope):
        trend_line = slope * series.index + intercept
        ax1.plot(series.index, trend_line, '--', color=colors[i], 
                 linewidth=2, alpha=0.8)

ax1.set_title('夏季 Albedo 时间序列及趋势线')
ax1.set_xlabel('年份')
ax1.set_ylabel('Albedo 均值')
ax1.legend()
ax1.grid(alpha=0.3)

# 子图2: 滑动标准差 (变异性)
ax2 = axes[0, 1]
window = 10  # 10年滑动窗口

for i, band in enumerate(df_all.columns):
    series = df_all[band]
    rolling_std = series.rolling(window=window, center=True).std()
    ax2.plot(rolling_std.index, rolling_std.values, 'o-', color=colors[i], 
             label=f'{band}', linewidth=1.5, markersize=3)

ax2.axvline(x=2005, color='red', linestyle='--', alpha=0.7, label='2005分界线')
ax2.set_title(f'{window}年滑动标准差 (变异性趋势)')
ax2.set_xlabel('年份')
ax2.set_ylabel('滑动标准差')
ax2.legend()
ax2.grid(alpha=0.3)

# 子图3: 分段趋势比较
ax3 = axes[1, 0]
periods = ['1950-1990', '1990-2010', '2010-2024']
x_pos = np.arange(len(periods))
num_bands = len(df_all.columns)
width = 0.8 / num_bands # 动态调整柱子宽度

for i, band in enumerate(df_all.columns):
    slopes = []
    for period in periods:
        start, end = map(int, period.split('-'))
        period_data = df_all[band][(df_all.index >= start) & (df_all.index <= end)]
        if len(period_data) >= 3:
            slope, _, _, _ = calculate_trend(period_data, period_data.index)
            slopes.append(slope * 10)  # 转换为每10年变化
        else:
            slopes.append(0)
    
    ax3.bar(x_pos - (width * (num_bands / 2 - 0.5)) + i * width, slopes, width, label=band, color=colors[i], alpha=0.8)

ax3.set_title('分段趋势比较 (每10年变化率)')
ax3.set_xlabel('时间段')
ax3.set_ylabel('Albedo 变化率 (/10年)')
ax3.set_xticks(x_pos)
ax3.set_xticklabels(periods)
ax3.legend()
ax3.grid(axis='y', alpha=0.3)
ax3.axhline(y=0, color='black', linestyle='-', linewidth=0.5)

# 子图4: 异常年份标识
ax4 = axes[1, 1]
for i, band in enumerate(df_all.columns):
    series = df_all[band]
    
    # 绘制标准化数据
    normalized = (series - series.mean()) / series.std()
    ax4.plot(series.index, normalized, 'o-', color=colors[i], 
             label=f'{band}', linewidth=1.5, markersize=4, alpha=0.7)
    
    # 标记异常值
    anomalies = detect_anomalies(series, threshold=2.0)
    for year in anomalies:
        if year in normalized.index:
            ax4.plot(year, normalized[year], 'x', color='red', markersize=10, markeredgewidth=3)

# 添加阈值线
ax4.axhline(y=2, color='red', linestyle='--', alpha=0.5, label='异常阈值 (±2σ)')
ax4.axhline(y=-2, color='red', linestyle='--', alpha=0.5)
ax4.axhline(y=0, color='black', linestyle='-', linewidth=0.5, alpha=0.5)

ax4.set_title('标准化 Albedo 与异常检测')
ax4.set_xlabel('年份')
ax4.set_ylabel('标准化 Albedo (Z-score)')
ax4.legend()
ax4.grid(alpha=0.3)

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()

# === STEP 8: 生成综合统计报告 ===
print("\n" + "="*80)
print("综合统计分析报告")
print("="*80)

detailed_stats = []

for band in df_all.columns:
    series = df_all[band]
    
    # 基础统计
    mean_all = series.mean()
    std_all = series.std()
    min_val = series.min()
    max_val = series.max()
    
    # 趋势分析
    slope_all, _, r2_all, p_all = calculate_trend(series, series.index)
    series_recent = series[series.index >= 2005]
    slope_recent, _, r2_recent, p_recent = calculate_trend(
        series_recent, series_recent.index
    )
    
    # 变异性分析
    early_std, recent_std, var_ratio, f_p = calculate_variability_change(series, 2005)
    
    # 异常检测
    anomalies = detect_anomalies(series, threshold=2.0)
    
    # 趋势显著性判断
    trend_significance_all = "显著" if not np.isnan(p_all) and p_all < 0.05 else "不显著"
    trend_significance_recent = "显著" if not np.isnan(p_recent) and p_recent < 0.05 else "不显著"
    
    # 变异性变化判断
    if not np.isnan(f_p) and f_p < 0.05:
        variability_change = "显著增加" if var_ratio > 1 else "显著减少"
    else:
        variability_change = "无显著变化"
    
    row = {
        '海拔带': band,
        '时间序列均值': f"{mean_all:.6f}",
        '时间序列标准差': f"{std_all:.6f}",
        '最小值(年份)': f"{min_val:.6f}({series.idxmin()})",
        '最大值(年份)': f"{max_val:.6f}({series.idxmax()})",
        '总体趋势(每年)': f"{slope_all:.8f}",
        '总体趋势(每10年)': f"{slope_all*10:.6f}",
        '总体趋势R²': f"{r2_all:.4f}",
        '总体趋势显著性': f"{trend_significance_all}(P={p_all:.4f})",
        '近20年趋势(每年)': f"{slope_recent:.8f}",
        '近20年趋势(每10年)': f"{slope_recent*10:.6f}",
        '近20年趋势R²': f"{r2_recent:.4f}",
        '近20年趋势显著性': f"{trend_significance_recent}(P={p_recent:.4f})",
        '前期变异性(2005前)': f"{early_std:.6f}",
        '近期变异性(2005后)': f"{recent_std:.6f}",
        '变异性比值': f"{var_ratio:.3f}",
        '变异性变化': f"{variability_change}(P={f_p:.4f})",
        '异常年份数': len(anomalies),
        '异常年份': ','.join(map(str, anomalies)) if anomalies else '无',
        '数据完整性': f"{(~series.isna()).sum()}/{len(series)}年"
    }
    
    detailed_stats.append(row)

# 转换为 DataFrame 并显示
df_detailed = pd.DataFrame(detailed_stats)

print("\n详细统计表格:")
print("-" * 120)
# 转置显示，便于阅读
print(df_detailed.set_index('海拔带').T.to_string())


# === STEP 9: 风险评估与预警 ===
print("\n" + "="*80)
print("风险评估与预警")
print("="*80)

for band in df_all.columns:
    series = df_all[band]
    
    # 计算风险指标
    risk_factors = []
    
    # 1. 趋势风险
    series_recent = series[series.index >= 2005]
    slope_recent, _, _, p_recent = calculate_trend(
        series_recent, series_recent.index
    )
    if not np.isnan(p_recent) and p_recent < 0.05 and slope_recent < -0.001:  # 显著下降趋势
        risk_factors.append(f"显著下降趋势({slope_recent*10:.4f}/10年)")
    
    # 2. 变异性风险
    _, _, var_ratio, f_p = calculate_variability_change(series, 2005)
    if not np.isnan(f_p) and f_p < 0.05 and var_ratio > 1.5:  # 变异性显著增加
        risk_factors.append(f"变异性显著增加({var_ratio:.2f}倍)")
    
    # 3. 异常频率风险
    recent_anomalies = [year for year in detect_anomalies(series) if year >= 2010]
    if len(recent_anomalies) >= 3:  # 近15年有3次以上异常
        risk_factors.append(f"近期异常频繁({len(recent_anomalies)}次)")
    
    # 4. 极值风险
    recent_series = series[series.index >= 2010]
    if not recent_series.empty and recent_series.min() == series.min():  # 历史最低值出现在近期
        risk_factors.append("历史最低值出现在近期")
    
    print(f"\n{band} 海拔带风险评估:")
    if risk_factors:
        print("  ⚠️  发现以下风险因子:")
        for factor in risk_factors:
            print(f"      • {factor}")
        
        # 综合风险等级
        risk_level = len(risk_factors)
        if risk_level >= 3:
            print("  🔴 综合风险等级: 高风险")
        elif risk_level == 2:
            print("  🟡 综合风险等级: 中等风险")
        else:
            print("  🟢 综合风险等级: 低风险")
    else:
        print("  ✅ 未发现显著风险因子")
        print("  🟢 综合风险等级: 低风险")

# === STEP 10: 保存结果 ===
print(f"\n{'='*60}")
print("分析完成")
print("="*60)

# 保存统计表格到 CSV
output_path = r"C:\greenland\albedo_elevation_analysis_results.csv"
try:
    df_detailed.to_csv(output_path, index=False, encoding='utf-8-sig')
    print(f"✅ 详细统计结果已保存至: {output_path}")
except Exception as e:
    print(f"📁 无法保存文件: {e}。请检查路径权限。")

print(f"✅ 分析完成! 共处理 {len(df_all.columns)} 个海拔带，{len(df_all)} 年数据")
print("📊 图表已显示，请查看控制台输出的统计表格进行进一步分析")
```

```python
# -*- coding: utf-8 -*-
import geopandas as gpd
import matplotlib.pyplot as plt
import numpy as np
from shapely.ops import unary_union, polygonize
from shapely.geometry import Polygon, MultiPolygon

# ===== 你的路径 =====
contour_shp  = r"C:\greenland\Greenland_shp\等间距等值线.shp"
boundary_shp = r"C:\greenland\Greenland_shp\Gboundry.shp"

# ===== 参数 =====
ELEV_FIELD  = "Contour"           # 高程字段
LEVELS_NEED = [1500, 2000, 2500]  # 需要分带的等高线

print("1) 读取数据 …")
contours = gpd.read_file(contour_shp)
boundary = gpd.read_file(boundary_shp)
contours[ELEV_FIELD] = contours[ELEV_FIELD].astype(float)

if contours.crs != boundary.crs:
    print(f"   重投影等高线 CRS {contours.crs} -> {boundary.crs}")
    contours = contours.to_crs(boundary.crs)

boundary_union = unary_union(boundary.geometry.buffer(0))

def area_ge_level(level):
    """基于给定等高线生成 >= level m 的区域（与边界相交）"""
    levels = contours[ELEV_FIELD].to_numpy()
    nearest = levels[np.argmin(np.abs(levels - level))]
    sel = contours[np.isclose(contours[ELEV_FIELD], nearest)]
    print(f"   目标 {level} m -> 使用 {nearest} m；线段数={len(sel)}")
    
    if sel.empty:
        raise RuntimeError(f"未找到 {level} m 等高线")
    
    merged_lines = unary_union(sel.geometry)
    polys = list(polygonize(merged_lines))
    
    if len(polys) == 0:
        raise RuntimeError(f"{level} m 等高线未形成闭合环，无法 polygonize")
    
    area = unary_union(polys).buffer(0)
    return area.intersection(boundary_union).buffer(0), float(nearest)

print("2) 生成 ≥1500 m、≥2000 m、≥2500 m 区域 …")
area_ge1500, used_1500 = area_ge_level(1500)
area_ge2000, used_2000 = area_ge_level(2000)
area_ge2500, used_2500 = area_ge_level(2500)

print("3) 计算四个分区 …")
band_highest = area_ge2500                                           # ≥2500
band_high    = area_ge2000.difference(area_ge2500).buffer(0)         # 2000–2500
band_mid     = area_ge1500.difference(area_ge2000).buffer(0)         # 1500–2000
band_low     = boundary_union.difference(area_ge1500).buffer(0)      # <1500

def to_gs(geom, crs):
    if geom.is_empty:
        return gpd.GeoSeries([], crs=crs)
    gs = gpd.GeoSeries([geom], crs=crs)
    return gs.explode(index_parts=False)

gdf_low     = to_gs(band_low,     boundary.crs)
gdf_mid     = to_gs(band_mid,     boundary.crs)
gdf_high    = to_gs(band_high,    boundary.crs)
gdf_highest = to_gs(band_highest, boundary.crs)

print(f"   <{used_1500:.0f} m: {len(gdf_low)} 多边形；"
      f"{used_1500:.0f}–{used_2000:.0f} m: {len(gdf_mid)}；"
      f"{used_2000:.0f}–{used_2500:.0f} m: {len(gdf_high)}；"
      f"≥{used_2500:.0f} m: {len(gdf_highest)}")

print("4) 绘图（四色分带） …")
fig, ax = plt.subplots(figsize=(6.4, 7.2))

# 低 -> 高 的顺序绘制，避免遮挡
gdf_low.plot(ax=ax,     color="#e67e22", edgecolor="black", linewidth=0.35)  # 橘色：<1500
gdf_mid.plot(ax=ax,     color="#f7dc6f", edgecolor="black", linewidth=0.35)  # 浅黄：1500–2000
gdf_high.plot(ax=ax,    color="#85c1e9", edgecolor="black", linewidth=0.35)  # 浅蓝：2000–2500
gdf_highest.plot(ax=ax, color="#76d7c4", edgecolor="black", linewidth=0.35)  # 浅绿：≥2500

boundary.boundary.plot(ax=ax, color="black", linewidth=0.6)

ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])
ax.set_title(f"Greenland elevation bands by {int(used_1500)}/{int(used_2000)}/{int(used_2500)} m contours", fontsize=11)
ax.set_frame_on(False)

import matplotlib.patches as mpatches
legend_patches = [
    mpatches.Patch(facecolor="#e67e22", edgecolor="black", label=f"< {int(used_1500)} m"),
    mpatches.Patch(facecolor="#f7dc6f", edgecolor="black", label=f"{int(used_1500)}–{int(used_2000)} m"),
    mpatches.Patch(facecolor="#85c1e9", edgecolor="black", label=f"{int(used_2000)}–{int(used_2500)} m"),
    mpatches.Patch(facecolor="#76d7c4", edgecolor="black", label=f"≥ {int(used_2500)} m"),
]
ax.legend(handles=legend_patches, loc="lower left", fontsize=9, frameon=False)

plt.tight_layout()
plt.show()
print("完成 ✅")
```

```python
# -*- coding: utf-8 -*-
import xarray as xr
import geopandas as gpd
import matplotlib.pyplot as plt
import rioxarray  # 确保已安装 pip install rioxarray geopandas shapely
import pandas as pd

# 文件路径
nc_path = r"C:\greenland\1995bu\Greenland_albedo1950-2024.nc"
shp_dir = r"C:\greenland\Greenland_shp\流域冰盖"   # 假设这里有 xiaorong.shp, dong.shp, xi.shp

# === 1. 读取 NetCDF 数据 ===
ds = xr.open_dataset(nc_path)
albedo = ds["albedo"]

# 只取夏季 (6,7,8 月)
albedo_summer = albedo.where(albedo["time.month"].isin([6, 7, 8]), drop=True)

# === 2. 读取分区 shapefile ===
regions = {
    "xb": gpd.read_file(f"{shp_dir}/西北流域.shp"),
    "x": gpd.read_file(f"{shp_dir}/西流域.shp"),
    "xn": gpd.read_file(f"{shp_dir}/西南流域.shp"),
}

# 结果保存
results = {}

# === 3. 每个分区计算夏季均值 ===
for name, gdf in regions.items():
    # CRS 确保一致
    if gdf.crs != "EPSG:4326":
        gdf = gdf.to_crs("EPSG:4326")
    geom = gdf.geometry.unary_union

    # rioxarray 裁剪
    albedo_clip = albedo_summer.rio.write_crs("EPSG:4326").rio.clip([geom], all_touched=True)

    # 按年份聚合
    df_mean = (
        albedo_clip.groupby("time.year").mean(dim=["time", "lat", "lon"]).to_dataframe().reset_index()
    )
    df_mean = df_mean.rename(columns={"albedo": name})
    results[name] = df_mean.set_index("year")[name]

# === 4. 合并所有分区结果 ===
df_all = pd.concat(results.values(), axis=1)
print(df_all.head())

# === 5. 画折线图 ===
plt.figure(figsize=(12, 6))
for name in results.keys():
    plt.plot(df_all.index, df_all[name], label=name, linewidth=1.5)

plt.title("1950–2024 格陵兰各分区夏季 (JJA) Albedo 均值", fontsize=14, fontweight="bold")
plt.xlabel("年份")
plt.ylabel("Albedo 均值")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# -*- coding: utf-8 -*-
import os, math, warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
import cartopy.crs as ccrs
from cartopy.io.shapereader import Reader
from cartopy.feature import ShapelyFeature

# ====== 路径与列名 ======
csv_path    = r"C:\greenland\merged_allfeature.csv"
col_station = "station"
col_lon     = "longitude"
col_lat     = "latitude"
col_year    = "year"
col_month   = "month"
col_albedo  = "albedo"

shp_xiaorong = r"C:\greenland\Greenland_shp\流域冰盖\new\xiaorong.shp"
shp_dong     = r"C:\greenland\Greenland_shp\流域冰盖\new\dong.shp"
shp_xi       = r"C:\greenland\Greenland_shp\流域冰盖\new\xi.shp"

# ====== 参数 ======
summer_months = {6, 7, 8}
year_min, year_max = 1995, 2024
extent = [-75, -10, 58, 85]  # [lon_min, lon_max, lat_min, lat_max]

# 分区阈值
TOP_FRAC = 0.25  # 最北25%进顶部
SIDE_SPLIT = "median"  # 左右用经度“median”（也可设为 "center" 用地图中心经度）

# 样式
line_width = 0.9
grid_alpha = 0.45
arrow_style = dict(arrowstyle="-|>", lw=0.9, color="0.28", shrinkA=0, shrinkB=4)

# ====== 读与汇总 ======
if not os.path.exists(csv_path):
    raise FileNotFoundError(csv_path)
df = pd.read_csv(csv_path, low_memory=False)

if (col_year not in df.columns) or (col_month not in df.columns):
    for cand in ["date","time","datetime","valid_time","timestamp","Date","Time","Datetime"]:
        if cand in df.columns:
            df[cand] = pd.to_datetime(df[cand], errors="coerce")
            df[col_year]  = df[cand].dt.year
            df[col_month] = df[cand].dt.month
            break

for c in [col_station, col_lon, col_lat, col_year, col_month, col_albedo]:
    if c not in df.columns:
        raise ValueError(f"缺少必要列：{c}")

coord_df = (df[[col_station, col_lon, col_lat]]
            .dropna()
            .drop_duplicates(subset=[col_station])
            .set_index(col_station))

df = df[df[col_month].isin(summer_months)]
df = df[(df[col_year] >= year_min) & (df[col_year] <= year_max)]
df = df.replace([np.inf, -np.inf], np.nan).dropna(subset=[col_albedo, col_station, col_year])

if df.empty:
    raise ValueError("筛选后夏季数据为空。")

jja = (df.groupby([col_station, col_year], as_index=False)[col_albedo]
         .mean()
         .rename(columns={col_albedo: "albedo_JJA_mean"}))
jja = jja[jja[col_station].isin(coord_df.index)]

pivot = jja.pivot(index=col_year, columns=col_station, values="albedo_JJA_mean").sort_index()
stations = [st for st in pivot.columns if pivot[st].count() > 0]
if not stations:
    raise ValueError("没有任何站点具备夏季均值。")

# ====== 分配到上/左/右（关键改动） ======
lon_min, lon_max, lat_min, lat_max = extent
lat_thr = lat_min + (lat_max - lat_min) * (1 - TOP_FRAC)  # 顶部阈值（越北越大）

# 左右分割经度：用站点经度“中位数”（更贴合站点分布）
station_lons = coord_df.loc[stations, col_lon].astype(float).values
if SIDE_SPLIT == "median":
    lon_split = float(np.median(station_lons))
else:  # "center"
    lon_split = (lon_min + lon_max) / 2.0

def side_of_station(lon, lat):
    if lat >= lat_thr:
        return "top"
    return "left" if lon <= lon_split else "right"

assign = {st: side_of_station(float(coord_df.loc[st, col_lon]),
                              float(coord_df.loc[st, col_lat]))
          for st in stations}

top_list  = [s for s in stations if assign[s] == "top"]
left_list = [s for s in stations if assign[s] == "left"]
right_list= [s for s in stations if assign[s] == "right"]

# 排序减少交叉：上=按经度西→东；左右=按纬度北→南
top_list.sort(key=lambda s: float(coord_df.loc[s, col_lon]))
left_list.sort(key=lambda s: -float(coord_df.loc[s, col_lat]))
right_list.sort(key=lambda s: -float(coord_df.loc[s, col_lat]))

# ====== 画布与主图 ======
fig = plt.figure(figsize=(16, 10), dpi=180)
map_rect = [0.24, 0.18, 0.52, 0.64]  # 中间地图
ax_map = plt.axes(map_rect, projection=ccrs.PlateCarree())
ax_map.set_extent(extent, crs=ccrs.PlateCarree())
ax_map.set_title("Greenland Regions & Stations  →  JJA Albedo per Station", pad=8)
ax_map.coastlines(linewidth=0.4, alpha=0.6)

def add_shp(ax, shp_path, fc, ec):
    if shp_path and os.path.exists(shp_path):
        feature = ShapelyFeature(Reader(shp_path).geometries(),
                                 ccrs.PlateCarree(), facecolor=fc,
                                 edgecolor=ec, linewidth=0.6, alpha=0.35)
        ax.add_feature(feature)

add_shp(ax_map, shp_xiaorong, "#5DADE2", "#2E86C1")
add_shp(ax_map, shp_dong,     "#F5B041", "#D68910")
add_shp(ax_map, shp_xi,       "#58D68D", "#239B56")

ax_map.scatter(coord_df.loc[stations, col_lon].values,
               coord_df.loc[stations, col_lat].values,
               s=18, c="k", marker="o",
               transform=ccrs.PlateCarree(), zorder=5)

# ====== 三边条带 ======
top_strip   = [map_rect[0], map_rect[1] + map_rect[3] + 0.03, map_rect[2], 0.16]
left_strip  = [map_rect[0] - 0.18, map_rect[1], 0.16, map_rect[3]]
right_strip = [map_rect[0] + map_rect[2] + 0.02, map_rect[1], 0.16, map_rect[3]]

def make_row_axes(fig, strip, n_panels):
    if n_panels == 0: return []
    x0, y0, w, h = strip
    pad_x, pad_y = 0.02, 0.16
    usable_w, usable_h = w*(1-2*pad_x), h*(1-2*pad_y)
    cell_w = usable_w / n_panels
    out=[]
    for i in range(n_panels):
        left = x0 + w*pad_x + i*cell_w
        bot  = y0 + h*pad_y
        ax   = fig.add_axes([left, bot, cell_w, usable_h])
        out.append((ax, (left+cell_w/2, bot+usable_h/2)))
    return out

def make_col_axes(fig, strip, n_panels):
    if n_panels == 0: return []
    x0, y0, w, h = strip
    pad_x, pad_y = 0.18, 0.04
    usable_w, usable_h = w*(1-2*pad_x), h*(1-2*pad_y)
    cell_h = usable_h / n_panels
    out=[]
    for i in range(n_panels):
        left = x0 + w*pad_x
        bot  = y0 + h*pad_y + (n_panels-1-i)*cell_h
        ax   = fig.add_axes([left, bot, usable_w, cell_h])
        out.append((ax, (left+usable_w/2, bot+cell_h/2)))
    return out

top_axes   = make_row_axes(fig, top_strip,   len(top_list))
left_axes  = make_col_axes(fig, left_strip,  len(left_list))
right_axes = make_col_axes(fig, right_strip, len(right_list))

# 统一Y轴范围
vals = pivot[stations].to_numpy().ravel()
vals = vals[~np.isnan(vals)]
if len(vals) >= 10:
    y_min = float(np.nanpercentile(vals, 1))
    y_max = float(np.nanpercentile(vals, 99))
    if y_min == y_max:
        y_min, y_max = float(np.nanmin(vals)), float(np.nanmax(vals))
else:
    y_min, y_max = float(np.nanmin(vals)), float(np.nanmax(vals))

def plot_series(ax, st, nbins=4):
    s = pivot[st].dropna()
    ax.plot(s.index.values, s.values, linewidth=line_width)
    ax.set_xlim(year_min, year_max)
    ax.set_ylim(y_min, y_max)
    ax.grid(True, linestyle="--", linewidth=0.35, alpha=grid_alpha)
    ax.xaxis.set_major_locator(MaxNLocator(integer=True, nbins=nbins))
    ax.yaxis.set_major_locator(MaxNLocator(nbins=4, prune='both'))
    ax.set_title(str(st), fontsize=8, pad=1.5)

# 顶部
for (st, (ax_ts, center)) in zip(top_list, top_axes):
    plot_series(ax_ts, st, nbins=5)
    lon, lat = float(coord_df.loc[st, col_lon]), float(coord_df.loc[st, col_lat])
    ax_map.annotate("", xy=(lon, lat), xycoords=ccrs.PlateCarree(),
                    xytext=center, textcoords=fig.transFigure,
                    arrowprops=arrow_style, zorder=10)

# 左/右
for (st, (ax_ts, center)) in zip(left_list, left_axes):
    plot_series(ax_ts, st, nbins=3)
    lon, lat = float(coord_df.loc[st, col_lon]), float(coord_df.loc[st, col_lat])
    ax_map.annotate("", xy=(lon, lat), xycoords=ccrs.PlateCarree(),
                    xytext=center, textcoords=fig.transFigure,
                    arrowprops=arrow_style, zorder=10)

for (st, (ax_ts, center)) in zip(right_list, right_axes):
    plot_series(ax_ts, st, nbins=3)
    lon, lat = float(coord_df.loc[st, col_lon]), float(coord_df.loc[st, col_lat])
    ax_map.annotate("", xy=(lon, lat), xycoords=ccrs.PlateCarree(),
                    xytext=center, textcoords=fig.transFigure,
                    arrowprops=arrow_style, zorder=10)

# 轴标签精简
for i,(ax_ts,_) in enumerate(left_axes):
    if i < len(left_axes)-1: ax_ts.set_xticklabels([])
for (ax_ts,_) in right_axes:
    ax_ts.set_yticklabels([])
for (ax_ts,_) in top_axes:
    ax_ts.set_yticklabels([])

fig.supxlabel("Year", y=0.06, fontsize=10)
fig.supylabel("Albedo (JJA mean)", x=0.02, fontsize=10)

# 保存
out_png = os.path.join(os.path.dirname(csv_path) or ".", "Greenland_map_arrow_panels_fix_left_right.png")
plt.savefig(out_png, dpi=220, bbox_inches="tight")
plt.close()
print(f"✅ 已保存：{out_png}")
```

```python
import pandas as pd
import geopandas
from shapely.geometry import Point
import os

def classify_greenland_points(csv_path, shp_folder_path, output_csv_path):
    """
    根据Greenland流域的shp文件对经纬度点进行分类。

    Args:
        csv_path (str): 包含经纬度坐标的CSV文件路径。
                        文件应包含 'latitude' 和 'longitude' 列。
        shp_folder_path (str): 包含流域shp文件的文件夹路径。
                                例如：西南流域.shp, S区.shp, N区.shp, NW区.shp
        output_csv_path (str): 分类结果输出的CSV文件路径。
    """
    # 1. 读取CSV文件
    try:
        df = pd.read_csv(csv_path)
    except FileNotFoundError:
        print(f"错误：未找到CSV文件：{csv_path}")
        return
    except Exception as e:
        print(f"读取CSV文件时发生错误：{e}")
        return

    # 确保 'latitude' 和 'longitude' 列存在
    if 'latitude' not in df.columns or 'longitude' not in df.columns:
        print("错误：CSV文件必须包含 'latitude' 和 'longitude' 列。")
        return

    # 创建GeoDataFrame
    # 假设CSV中的坐标是WGS84 (EPSG:4326)
    gdf_points = geopandas.GeoDataFrame(
        df, geometry=geopandas.points_from_xy(df.longitude, df.latitude), crs="EPSG:4326"
    )

    # 初始化 'region' 列
    df['region'] = '未分类'

    # 2. 读取所有的shp文件并存储到字典中
    region_shapefiles = {}
    # 关键修改：定义中文区域名称与代码中使用的英文缩写的映射
    # 请根据你的实际需求调整这里的映射关系
    # 例如：如果你希望最终分类结果显示 '西南' 而不是 'SW'，那么字典的键就是 '西南'
    # 但为了兼容你之前说的 SW, S, N, NW 区，我们让字典的键保持英文缩写，值是实际的文件名。
    
    # 这里的键 (SW, S, N, NW) 会是最终分类到CSV文件中的 'region' 列的值
    # 这里的值 (西南流域.shp, 南流域.shp, 北流域.shp, 西北流域.shp) 是实际的文件名
    region_mapping = {
        'SW': '西南流域.shp',
        'S': '南流域.shp',
        'N': '北流域.shp',
        'NW': '西北流域.shp',
        # 根据你提供的文件列表，可能还有其他区域需要分类，例如：
        'NE': '东北流域.shp',
        'SE': '东南流域.shp',
        'E': '东流域.shp',
        'W': '西流域.shp',
    }

    # 遍历映射来加载Shapefile
    for region_code, shp_base_name in region_mapping.items():
        shp_file_path = os.path.join(shp_folder_path, shp_base_name)
        try:
            # 读取shp文件，并将其投影到与点相同的CRS
            region_shapefiles[region_code] = geopandas.read_file(shp_file_path).to_crs(gdf_points.crs)
            print(f"成功加载shp文件: {shp_file_path} (对应区域: {region_code})")
        except FileNotFoundError:
            print(f"警告：未找到shp文件：{shp_file_path}。请确认文件是否存在，将跳过此区域的分类。")
        except Exception as e:
            print(f"读取shp文件 {shp_file_path} (对应区域: {region_code}) 时发生错误：{e}。将跳过此区域的分类。")

    # 3. 对每个点进行分类
    for index, point_row in gdf_points.iterrows():
        point_geometry = point_row.geometry
        assigned = False
        # 遍历已成功加载的区域Shapefile
        # 注意：这里遍历的顺序决定了当一个点落在多个区域边界时，它会被分配给哪个区域
        # 默认按照字典的定义顺序 (SW, S, N, NW, NE, SE, E, W...)
        for region_code, region_gdf in region_shapefiles.items():
            # 遍历shp文件中的所有几何体（多边形）
            for _, polygon_row in region_gdf.iterrows():
                if point_geometry.within(polygon_row.geometry):
                    df.loc[index, 'region'] = region_code # 分类到对应的英文区域代码
                    assigned = True
                    break # 找到所属区域后跳出当前shp文件的多边形遍历
            if assigned:
                break # 找到所属区域后跳出区域遍历

    # 4. 保存分类结果到新的CSV文件
    try:
        df.to_csv(output_csv_path, index=False)
        print(f"\n分类完成！结果已保存到：{output_csv_path}")
    except Exception as e:
        print(f"保存结果CSV文件时发生错误：{e}")

# --- 使用示例 ---
if __name__ == "__main__":
    csv_file = r"F:\TUB\greenlandProject\ne_10m_admin_0_countries\daqucsv.csv"
    # 请确保这里的路径是你实际的文件夹路径
    shp_folder = r"E:\TUB\greenland\相关数据\流域冰盖"
    output_classified_csv = r"F:\TUB\greenlandProject\ne_10m_admin_0_countries\classified_points.csv"

    classify_greenland_points(csv_file, shp_folder, output_classified_csv)
```

```python
#所有指定区域的反照率统计任务已完成------------"C:\Users\11823\era5land\greenland_SW_trend.csv"----------------------八个分区和------非冰盖区----------------------
import pandas as pd
import os
import numpy as np

def calculate_albedo_statistics(albedo_folder, classified_points_file, output_file, target_region):
    """
    Calculates monthly albedo statistics (mean, min, max, std, median, Q1, Q3, count)
    for a specified region over a range of years and months.
    Includes print statements to show progress.

    Args:
        albedo_folder (str): Path to the folder containing monthly albedo CSV files.
        classified_points_file (str): Path to the CSV file with classified points and regions.
        output_file (str): Path to the output CSV file to store the calculated trends.
        target_region (str): The specific region to calculate statistics for.
    """

    print(f"\n{'='*70}")
    print(f"--- 开始处理区域: '{target_region}' 的反照率数据 ---")
    print(f"分类点文件: {classified_points_file}")
    print(f"反照率数据文件夹: {albedo_folder}")
    print(f"输出文件: {output_file}\n")

    # Load the classified points data
    try:
        print(f"正在加载分类点文件: {classified_points_file}...")
        classified_points_df = pd.read_csv(classified_points_file)
        # Filter for the target region only once
        target_region_points = classified_points_df[
            classified_points_df['region'] == target_region
        ][['latitude', 'longitude']]
        # Use a tuple of (latitude, longitude) for faster lookup
        target_region_coords = set(
            tuple(row) for row in target_region_points.to_numpy()
        )
        print(f"已加载 {len(target_region_points)} 个 '{target_region}' 区域的参考点位。\n")
        
        if len(target_region_points) == 0:
            print(f"警告: 在 classified_points.csv 中未找到 '{target_region}' 区域的任何点位。跳过此区域。")
            return # Exit the function if no points are found for the region

    except FileNotFoundError:
        print(f"错误: 分类点文件未找到，请检查路径: {classified_points_file}")
        return
    except KeyError:
        print(
            "错误: classified_points.csv 文件中缺少 'latitude', 'longitude' 或 'region' 列。"
        )
        return
    except Exception as e:
        print(f"加载分类点文件时发生意外错误: {e}")
        return

    # List to store monthly statistics
    all_monthly_stats = []

    # Get a sorted list of albedo files
    albedo_files = sorted(
        [f for f in os.listdir(albedo_folder) if f.endswith('Greenland_albedo.csv')]
    )

    if not albedo_files:
        print(f"错误: 在 {albedo_folder} 文件夹中未找到任何反照率CSV文件。请检查文件夹路径和文件命名规则。")
        return

    total_files = len(albedo_files)
    print(f"共找到 {total_files} 个反照率数据文件，开始逐月处理 '{target_region}' 区域...\n")

    for i, filename in enumerate(albedo_files):
        file_path = os.path.join(albedo_folder, filename)
        
        # Extract year and month from the filename (e.g., 195004Greenland_albedo.csv)
        try:
            year_month_str = filename[:6]
            year = int(year_month_str[:4])
            month = int(year_month_str[4:])
            time_label = f"{year}/{month:02d}"  # Format as YYYY/MM
            print(f"[{i+1}/{total_files}] 正在处理文件: {filename} ({time_label})")
        except ValueError:
            print(f"跳过格式错误的文件名: {filename}。无法提取年份和月份。")
            continue

        try:
            current_albedo_df = pd.read_csv(file_path)
        except FileNotFoundError:
            print(f"错误: 反照率文件未找到: {file_path}")
            continue
        except Exception as e:
            print(f"读取文件 {file_path} 时发生错误: {e}")
            continue

        # Filter albedo data for the target region's points
        # Create a temporary column for tuple (latitude, longitude) for efficient filtering
        current_albedo_df['coord_tuple'] = list(
            zip(current_albedo_df['latitude'], current_albedo_df['longitude'])
        )
        
        # Filter rows where the coordinate tuple is in our set of target region coordinates
        n_region_albedo = current_albedo_df[
            current_albedo_df['coord_tuple'].isin(target_region_coords)
        ]['albedo']

        num_filtered_points = len(n_region_albedo)
        print(f"  筛选出 '{target_region}' 区域的点有 {num_filtered_points} 个。")

        if not n_region_albedo.empty:
            stats = {
                'time': time_label,
                'albedo_mean': n_region_albedo.mean(),
                'albedo_median': n_region_albedo.median(),
                'albedo_std': n_region_albedo.std(),
                'albedo_min': n_region_albedo.min(),
                'albedo_max': n_region_albedo.max(),
                'albedo_q1': n_region_albedo.quantile(0.25),
                'albedo_q3': n_region_albedo.quantile(0.75),
                'albedo_count': n_region_albedo.count()
            }
            all_monthly_stats.append(stats)
            print(f"  {time_label} 的统计量已计算完成。")
        else:
            print(f"  在文件 {filename} 中未找到 '{target_region}' 区域的反照率数据。跳过此月。")
        print("-" * 50) # Add a separator for clarity

    # Create a DataFrame from the collected statistics
    if all_monthly_stats:
        output_df = pd.DataFrame(all_monthly_stats)
        
        # Ensure chronological order by sorting
        output_df['year_temp'] = output_df['time'].apply(lambda x: int(x.split('/')[0]))
        output_df['month_temp'] = output_df['time'].apply(lambda x: int(x.split('/')[1]))
        output_df = output_df.sort_values(by=['year_temp', 'month_temp']).drop(columns=['year_temp', 'month_temp'])
        
        # Save the DataFrame to a CSV file
        output_df.to_csv(output_file, index=False)
        print(f"\n--- 区域 '{target_region}' 的数据处理完毕 ---")
        print(f"反照率统计结果已成功保存到: {output_file}")
    else:
        print(f"\n--- 区域 '{target_region}' 处理完成 ---")
        print(f"未能生成任何统计数据。请检查该区域是否存在数据。")
    print(f"\n{'='*70}\n")


# --- 配置参数 ---
albedo_data_folder = "F:\\TUB\\greenlandProject\\ne_10m_admin_0_countries\\greenlandAlbedo"
classified_points_file_path = (
    "F:\\TUB\\greenlandProject\\ne_10m_admin_0_countries\\classified_points.csv"
)

# Define all the regions you want to process
all_regions = ['N', 'E', 'NE', 'NW', 'S', 'SE', 'SW', 'W']

# --- 循环处理每个区域 ---
for region in all_regions:
    # Construct the output file name for the current region
    output_csv_path = f"greenland_{region}_trend.csv"
    
    # Call the function for each region
    calculate_albedo_statistics(
        albedo_data_folder,
        classified_points_file_path,
        output_csv_path,
        target_region=region,
    )

print("\n--- 所有指定区域的反照率统计任务已完成 ---")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

def plot_albedo_trends(data_folder, output_folder):
    """
    Reads albedo trend CSV files from a specified folder,
    plots the 'albedo_mean' over 'time' for each file,
    and saves each plot to a specified output folder.

    Args:
        data_folder (str): Path to the folder containing the trend CSV files.
        output_folder (str): Path to the folder where plots will be saved.
    """
    print(f"--- 开始绘制反照率时间变化图 ---")
    print(f"数据文件夹: {data_folder}")
    print(f"图表将保存到: {output_folder}\n")

    # Ensure the output folder exists
    os.makedirs(output_folder, exist_ok=True)

    # Get a list of all trend CSV files in the specified folder
    # Filter for files that start with 'greenland_' and end with '_trend.csv'
    trend_files = sorted(
        [f for f in os.listdir(data_folder) if f.startswith('greenland_') and f.endswith('_trend.csv')]
    )

    if not trend_files:
        print(f"错误: 在 '{data_folder}' 文件夹中未找到任何符合 'greenland_X_trend.csv' 命名模式的CSV文件。")
        print("请确保文件已存在且命名正确。")
        return

    print(f"共找到 {len(trend_files)} 个趋势文件，将逐一绘制并保存。\n")

    for i, filename in enumerate(trend_files):
        file_path = os.path.join(data_folder, filename)
        plot_filename = filename.replace(".csv", "_albedo_trend.png")
        output_file_path = os.path.join(output_folder, plot_filename)
        
        print(f"[{i+1}/{len(trend_files)}] 正在处理文件: {filename}")

        try:
            # Read the CSV file
            df = pd.read_csv(file_path)

            # Ensure 'time' column is in datetime format for proper plotting
            df['time'] = pd.to_datetime(df['time'])

            # Create the plot
            plt.figure(figsize=(12, 6)) # Set figure size for better readability
            plt.plot(df['time'], df['albedo_mean'], marker='o', markersize=2, linestyle='-')

            # Set plot title and labels
            plt.title(f'{filename.replace(".csv", "")} - Albedo Mean Over Time', fontsize=16)
            plt.xlabel('时间', fontsize=12)
            plt.ylabel('反照率均值 (Albedo Mean)', fontsize=12)
            plt.grid(True, linestyle='--', alpha=0.7) # Add a grid for easier reading

            # Improve x-axis date formatting
            plt.gcf().autofmt_xdate() # Automatically format x-axis labels for dates

            # Adjust layout to prevent labels from overlapping
            plt.tight_layout() 

            # Save the plot to the specified output folder
            plt.savefig(output_file_path)
            print(f"  图表已保存到: {output_file_path}")
            plt.close() # Close the plot to free up memory
            
        except FileNotFoundError:
            print(f"错误: 未找到文件 {file_path}。跳过。")
        except KeyError as e:
            print(f"错误: 文件 {filename} 中缺少必要的列（'time' 或 'albedo_mean'）。详细信息: {e}")
        except Exception as e:
            print(f"处理文件 {filename} 时发生未知错误: {e}")
        print("-" * 50) # Separator for clarity

    print("\n--- 所有图表绘制完成 ---")

# --- Configuration ---
# Set the path to the folder containing your trend CSV files
era5land_folder = "C:\\Users\\11823\\era5land" 
# Set the path to the folder where you want to save the plots
output_plots_folder = "C:\\Users\\11823\\era5land" # Saving plots in the same folder

# --- Run the plotting function ---
plot_albedo_trends(era5land_folder, output_plots_folder)
```

```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point
import numpy as np

# === 路径设置 ===
csv_path = r"F:\TUB\greenlandProject\ne_10m_admin_0_countries\GreenlandBoundry.csv"
shp_path = r"E:\TUB\greenland\Greenland_shp\Gboundry.shp"

# === 加载CSV并转换为GeoDataFrame ===
csv_df = pd.read_csv(csv_path)
geometry = [Point(xy) for xy in zip(csv_df['longitude'], csv_df['latitude'])]
csv_gdf = gpd.GeoDataFrame(csv_df, geometry=geometry, crs="EPSG:4326")

# === 读取Shapefile ===
shp_gdf = gpd.read_file(shp_path)
shp_gdf = shp_gdf.to_crs("EPSG:4326")  # 确保坐标系一致

# === 检查哪些CSV点在shapefile范围内 ===
csv_gdf['within_shp'] = csv_gdf.geometry.within(shp_gdf.unary_union)

# === 输出不在shapefile范围内的点（可选）===
outside_points = csv_gdf[~csv_gdf['within_shp']]
print(f"❗ 不在Shapefile内的CSV点数量: {len(outside_points)}")

# === 生成Shapefile覆盖范围内的所有0.1°网格点 ===
minx, miny, maxx, maxy = shp_gdf.total_bounds
lons = np.arange(np.floor(minx*10)/10, np.ceil(maxx*10)/10 + 0.01, 0.1)
lats = np.arange(np.floor(miny*10)/10, np.ceil(maxy*10)/10 + 0.01, 0.1)

grid_points = [Point(lon, lat) for lon in lons for lat in lats]
grid_gdf = gpd.GeoDataFrame(geometry=grid_points, crs="EPSG:4326")

# 筛选在shapefile范围内的网格点
grid_in_shp = grid_gdf[grid_gdf.geometry.within(shp_gdf.unary_union)]

# === 找出shapefile范围内但不在CSV中的点 ===
existing_points_set = set((round(p.x, 6), round(p.y, 6)) for p in csv_gdf.geometry)
new_points = [(pt.x, pt.y) for pt in grid_in_shp.geometry if (round(pt.x, 6), round(pt.y, 6)) not in existing_points_set]

# === 输出结果 ===
new_points_df = pd.DataFrame(new_points, columns=['longitude', 'latitude'])
new_points_df.to_csv("missing_grid_points.csv", index=False)

print(f"✅ 需要补充的网格点数量: {len(new_points_df)}")
print("➡️ 已保存至 missing_grid_points.csv")
```

```python
#遍历csv文件，补齐boundry中确实的1453个点-------------------1995bu----------------
import pandas as pd
import numpy as np
from scipy.spatial import cKDTree
from pathlib import Path

# 路径设置
input_dir = Path(r"F:\TUB\greenlandProject\ne_10m_admin_0_countries\greenlandAlbedo\1995prediction")
output_dir = Path(r"F:\TUB\greenlandProject\ne_10m_admin_0_countries\greenlandAlbedo\1995bu")
output_dir.mkdir(parents=True, exist_ok=True)

# 读取参考格网点，只保留经纬度
reference_path = Path(r"F:\TUB\greenlandProject\ne_10m_admin_0_countries\GreenlandBoundryBu.csv")
df_ref = pd.read_csv(reference_path)[['latitude', 'longitude']].drop_duplicates()

# 总计新增
total_added = 0

# 遍历每个albedo预测文件
for file in input_dir.glob("*GMonthlyalbedo.csv"):
    try:
        print(f"🔄 正在处理: {file.name}")
        df_partial = pd.read_csv(file)[['latitude', 'longitude', 'altitude_last_valid', 'albedo']]

        # 找出缺失点（ref中有，文件中没有）
        df_merge = df_ref.merge(df_partial, on=['latitude', 'longitude'], how='left', indicator=True)
        df_missing = df_merge[df_merge['_merge'] == 'left_only'][['latitude', 'longitude']]

        added_count = len(df_missing)

        if added_count > 0:
            # 最近邻插值albedo
            tree = cKDTree(df_partial[['latitude', 'longitude']].values)
            _, indices = tree.query(df_missing[['latitude', 'longitude']].values, k=1)
            df_missing['albedo'] = df_partial.iloc[indices]['albedo'].values

            # 保留空高程列
            df_missing['altitude_last_valid'] = np.nan

            # 调整列顺序
            df_missing = df_missing[['latitude', 'longitude', 'altitude_last_valid', 'albedo']]

            # 合并
            df_result = pd.concat([df_partial, df_missing], ignore_index=True)
        else:
            df_result = df_partial.copy()

        df_result = df_result.sort_values(by=['latitude', 'longitude'])
        output_path = output_dir / file.name
        df_result.to_csv(output_path, index=False)

        print(f"✅ 补完: {file.name}，新增 {added_count} 行")
        total_added += added_count

    except Exception as e:
        print(f"❌ 错误: {file.name} -> {e}")

print(f"\n📊 所有文件总共补了 {total_added} 行 albedo 数据")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress
import pymannkendall as mk
from statsmodels.tsa.seasonal import STL
from sklearn.cluster import AgglomerativeClustering
from sklearn.preprocessing import StandardScaler
from matplotlib.font_manager import FontProperties
import warnings

warnings.filterwarnings("ignore")

# 中文字体配置
font_path = r"C:\Windows\Fonts\msyh.ttc"
font_prop = FontProperties(fname=font_path)
plt.rcParams['axes.unicode_minus'] = False

# 读取数据
data_path = r"F:\TUB\greenlandProject\ne_10m_admin_0_countries\greenlandAlbedo\1995bu\mean.csv"
df = pd.read_csv(data_path)

# 只保留4-9月数据
target_months = [4,5,6,7,8,9]
df = df[df['month'].isin(target_months)]

# 计算夏季（6-8月）和年均（4-9月）
summer_df = df[df['month'].isin([6,7,8])].groupby('year')['albedo'].mean().reset_index(name='summer_mean')
yearly_df = df.groupby('year')['albedo'].mean().reset_index(name='yearly_mean')

# 宽格式
month_pivot = df.pivot(index='year', columns='month', values='albedo')

# 1. 每月线性趋势简洁输出
print("📌 每月线性趋势:")
monthly_trends = {}
for month in target_months:
    y = month_pivot[month].values
    x = month_pivot.index.values
    slope, _, r_value, p_value, _ = linregress(x, y)
    monthly_trends[month] = slope
    trend = "上升" if slope > 0 else ("下降" if slope < 0 else "无明显趋势")
    print(f"  月份 {month}: 斜率={slope:.5f} ({trend})，R²={r_value**2:.3f}, p={p_value:.3f}")

# 2. 非参数检验和Theil-Sen及bootstrap置信区间
def bootstrap_sen_slope(x, n_bootstrap=1000):
    n = len(x)
    slopes = []
    for _ in range(n_bootstrap):
        sample = np.random.choice(x, n, replace=True)
        years = np.arange(n)
        slope, _, _, _, _ = linregress(years, sample)
        slopes.append(slope)
    lower = np.percentile(slopes, 2.5)
    upper = np.percentile(slopes, 97.5)
    median = np.median(slopes)
    return median, lower, upper

print("\n📌 非参数检验与Theil-Sen斜率估计:")
for label, series in [("夏季均值", summer_df['summer_mean']),
                      ("年均", yearly_df['yearly_mean'])]:
    result = mk.original_test(series)
    slope = result.slope if hasattr(result, 'slope') else np.nan
    p_value = result.p
    trend = result.trend
    med, low, up = bootstrap_sen_slope(series.values)
    print(f"  {label}趋势: {trend}，p={p_value:.4f}，Sen斜率={slope:.5f}")
    print(f"    Bootstrap中位数斜率={med:.5f}，95%置信区间=({low:.5f}, {up:.5f})")

# 3. 5年滑动窗口趋势分析（夏季均值示范）
def sliding_window_trends(x, y, window=5):
    slopes = []
    for i in range(len(x) - window + 1):
        xs = x[i:i+window]
        ys = y[i:i+window]
        slope, _, _, _, _ = linregress(xs, ys)
        slopes.append(slope)
    return slopes

years = summer_df['year'].values
slopes = sliding_window_trends(years, summer_df['summer_mean'].values)
print("\n📌 5年滑动窗口趋势（夏季均值）:")
for i, slope in enumerate(slopes):
    trend = "上升" if slope > 0 else ("下降" if slope < 0 else "无明显趋势")
    print(f"  {years[i]}-{years[i+4]}: 斜率={slope:.5f} ({trend})")

plt.figure(figsize=(10,5))
plt.plot(years[4:], slopes, marker='o')
plt.axhline(0, color='gray', linestyle='--')
plt.title('夏季平均反照率 5年滑动窗口趋势斜率', fontproperties=font_prop)
plt.xlabel('年份', fontproperties=font_prop)
plt.ylabel('斜率', fontproperties=font_prop)
plt.grid(True)
plt.tight_layout()
plt.show()

# 4. STL季节分解（周期6，对4-9月）
monthly_mean = df.groupby(['year','month'])['albedo'].mean().reset_index()
monthly_mean['date'] = pd.to_datetime(monthly_mean['year'].astype(str) + '-' + monthly_mean['month'].astype(str) + '-15')
ts = monthly_mean.set_index('date')['albedo'].sort_index()

stl = STL(ts, period=6)
res = stl.fit()

plt.figure(figsize=(12,9))
plt.subplot(4,1,1)
plt.plot(res.observed, label='观测值')
plt.legend(prop=font_prop)
plt.subplot(4,1,2)
plt.plot(res.trend, label='趋势', color='orange')
plt.legend(prop=font_prop)
plt.subplot(4,1,3)
plt.plot(res.seasonal, label='季节性', color='green')
plt.legend(prop=font_prop)
plt.subplot(4,1,4)
plt.plot(res.resid, label='残差', color='red')
plt.legend(prop=font_prop)
plt.suptitle('STL季节分解', fontproperties=font_prop)
plt.tight_layout()
plt.show()

# 5. 趋势分组聚类（基于每月线性斜率）
trend_features = np.array([monthly_trends[m] for m in target_months]).reshape(-1,1)
scaler = StandardScaler()
trend_features_scaled = scaler.fit_transform(trend_features)
clustering = AgglomerativeClustering(n_clusters=2).fit(trend_features_scaled)

plt.figure(figsize=(8,4))
colors = ['C0' if c==0 else 'C1' for c in clustering.labels_]
plt.bar([str(m) for m in target_months], trend_features.flatten(), color=colors)
plt.xlabel('月份', fontproperties=font_prop)
plt.ylabel('趋势斜率', fontproperties=font_prop)
plt.title('月份趋势斜率聚类', fontproperties=font_prop)
plt.tight_layout()
plt.show()

print("\n📌 聚类结果:")
for m, c in zip(target_months, clustering.labels_):
    print(f"  月份 {m} 属于类别 {c}")

# 6. 绘制整体年均和夏季均值趋势
plt.figure(figsize=(10,6))
plt.plot(yearly_df['year'], yearly_df['yearly_mean'], label='4-9月年均')
plt.plot(summer_df['year'], summer_df['summer_mean'], label='6-8月夏季均值')
plt.xlabel('年份', fontproperties=font_prop)
plt.ylabel('Albedo', fontproperties=font_prop)
plt.title('格陵兰反照率时间序列', fontproperties=font_prop)
plt.legend(prop=font_prop)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# -------------------------------空间趋势计算--------------------------每个格网点--------------1950-2-24-----------------Step 0: 导入库
import os
import pandas as pd
import numpy as np
from math import sqrt
from collections import defaultdict
import matplotlib.pyplot as plt
from tqdm.notebook import tqdm  # 为Jupyter优化显示

# Step 1: 设置路径与文件列表
folder = r"F:\TUB\greenlandProject\ne_10m_admin_0_countries\greenlandAlbedo\1995bu"
all_files = sorted([f for f in os.listdir(folder) if f.endswith("albedo.csv")])
print(f"共找到 {len(all_files)} 个文件")

# Step 2: 读取所有数据（按时间排序）
time_list = []
albedo_dict = defaultdict(list)

for file in tqdm(all_files, desc="读取数据"):
    filepath = os.path.join(folder, file)
    df = pd.read_csv(filepath)

    # 计算浮点型时间：例如 202409 -> 2024.6667
    time = int(file[:6])
    year = time // 100
    month = time % 100
    time_float = year + (month - 1) / 12
    time_list.append(time_float)

    for _, row in df.iterrows():
        lat, lon, albedo = row['latitude'], row['longitude'], row['albedo']
        if not np.isnan(albedo):
            albedo_dict[(lat, lon)].append((time_float, albedo))

print(f"共收集到 {len(albedo_dict)} 个经纬度点")

# Step 3: 趋势函数（Mann-Kendall + Theil-Sen）
def mk_trend(time_series):
    time_series.sort()  # 按时间排序
    x = [v for _, v in time_series]
    n = len(x)
    if n < 5:
        return np.nan, np.nan

    # Mann-Kendall
    s = 0
    for m in range(n-1):
        for k in range(m+1, n):
            if x[k] > x[m]:
                s += 1
            elif x[k] < x[m]:
                s -= 1
    var_s = n*(n-1)*(2*n+5)/18.0
    if s > 0:
        z = (s - 1) / sqrt(var_s)
    elif s < 0:
        z = (s + 1) / sqrt(var_s)
    else:
        z = 0
    z_abs = abs(z)

    # Theil-Sen 斜率
    slopes = []
    for m in range(n-1):
        for k in range(m+1, n):
            delta = (x[k] - x[m]) / (k - m)
            slopes.append(delta)
    slope = np.median(slopes)
    return slope, z_abs

# Step 4: 遍历所有经纬度点，计算趋势
results = []
for (lat, lon), series in tqdm(albedo_dict.items(), desc="趋势分析"):
    slope, z = mk_trend(series)
    slope_decadal = slope * 10 if slope is not np.nan else np.nan
    sig = 1 if z >= 1.96 else 0
    results.append((lat, lon, slope_decadal, z, sig))

# Step 5: 输出结果 CSV
df_out = pd.DataFrame(results, columns=["latitude", "longitude", "slope_per_decade", "z_value", "significant"])
output_path = os.path.join(folder, "Greenland_albedo_trend_results.csv")
df_out.to_csv(output_path, index=False)
print(f"趋势结果已保存为：{output_path}")
```

```python
import os
import pandas as pd

# 设置文件夹路径
folder = r"F:\TUB\greenlandProject\ne_10m_admin_0_countries\greenlandAlbedo\1995bu"
output_file = os.path.join(folder, "April_to_September_Mean.csv")

# 用于存储所有数据 {(year, month): mean_albedo}
albedo_dict = {}

# 遍历所有相关文件
for filename in os.listdir(folder):
    if filename.endswith("GMonthlyalbedo.csv"):
        yyyymm = filename[:6]
        if len(yyyymm) == 6 and yyyymm.isdigit():
            year = int(yyyymm[:4])
            month = int(yyyymm[4:6])
            if 4 <= month <= 9:  # 只处理4–9月
                filepath = os.path.join(folder, filename)
                try:
                    df = pd.read_csv(filepath)
                    if 'albedo' in df.columns:
                        mean_albedo = df['albedo'].mean()
                        albedo_dict[(year, month)] = mean_albedo
                    else:
                        print(f"'albedo'列在 {filename} 中不存在，已跳过。")
                except Exception as e:
                    print(f"读取文件 {filename} 出错：{e}")

# 组织成DataFrame：行为年份，列为4–9月
data = {}
for year, month in sorted(albedo_dict.keys()):
    if year not in data:
        data[year] = {}
    data[year][f"{month}albedo"] = albedo_dict[(year, month)]

# 创建DataFrame并排序列
result_df = pd.DataFrame.from_dict(data, orient='index')
result_df.index.name = 'time'
result_df = result_df[['4albedo', '5albedo', '6albedo', '7albedo', '8albedo', '9albedo']]  # 保证列顺序
result_df.reset_index(inplace=True)

# 保存CSV
result_df.to_csv(output_file, index=False, float_format='%.8f')

print(f"每年4–9月albedo均值已保存至：{output_file}")
```

```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import os

# 路径设置
folder = r"F:\TUB\greenlandProject\ne_10m_admin_0_countries\greenlandAlbedo\1995bu"
input_file = os.path.join(folder, "April_to_September_Mean.csv")

# 读取数据
df = pd.read_csv(input_file)
month_cols = ['4albedo', '5albedo', '6albedo', '7albedo', '8albedo', '9albedo']
X = df[month_cols].copy()

# 标准化
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# PCA
pca = PCA()
scores = pca.fit_transform(X_scaled)

# 输出结果
print("Explained variance ratio (各主成分方差贡献率):")
print(pca.explained_variance_ratio_)

print("\nPCA components (各主成分的月权重):")
components_df = pd.DataFrame(pca.components_, columns=month_cols)
print(components_df)

print("\nPCA scores (每年对应各主成分得分):")
scores_df = pd.DataFrame(scores, columns=[f"PC{i+1}" for i in range(scores.shape[1])])
scores_df.insert(0, 'Year', df['time'])
print(scores_df)
```

```python
import pandas as pd
import os

# 路径
file_path = r"E:\TUB\greenland\1995bu\April_to_September_Mean.csv"

# 读取数据
df = pd.read_csv(file_path)

# 月份列
month_cols = ['4albedo', '5albedo', '6albedo', '7albedo', '8albedo', '9albedo']

# 计算各月与year列的相关系数
corr_dict = {}
for month in month_cols:
    corr = df[month].corr(df['year'])
    corr_dict[month] = corr

# 按相关系数大小排序
corr_sorted = sorted(corr_dict.items(), key=lambda x: abs(x[1]), reverse=True)

# 打印结果
print("各个月份与年均值(year)的相关系数（绝对值由大到小）：")
for month, corr in corr_sorted:
    print(f"{month}: {corr:.4f}")
```

```python
import pandas as pd

file_path = r"E:\TUB\greenland\1995bu\April_to_September_Mean.csv"
df = pd.read_csv(file_path)
month_cols = ['4albedo', '5albedo', '6albedo', '7albedo', '8albedo', '9albedo']

std_dict = {m: df[m].std() for m in month_cols}
std_sorted = sorted(std_dict.items(), key=lambda x: x[1], reverse=True)

print("各月份albedo年际标准差（变化幅度）：")
for month, std in std_sorted:
    print(f"{month}: {std:.4f}")
```

```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import linregress

# === 1. 路径配置 ===
file_path = r"F:\TUB\greenlandProject\ERA5LAND\greenland_tcc_full.nc"

# === 2. 读取和处理 tcc（仅提取每年7月） ===
ds = xr.open_dataset(file_path)
tcc = ds['tcc']
tcc = tcc.sel(valid_time=slice('1950-01', '2024-12'))
tcc_sel = tcc.sel(valid_time=tcc['valid_time.month'] == 7).groupby('valid_time.year').mean('valid_time')

# === 3. (a) 每年7月的平均云量空间图 ===
mean_tcc = tcc_sel.mean(dim='year')
plt.figure(figsize=(10, 8))
mean_tcc.plot(cmap='Blues', cbar_kwargs={'label': 'Mean Cloud Cover (Fraction)'})
plt.title('Greenland Average TCC (July, 1950–2024)')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.grid(False)
plt.savefig('Greenland_Avg_TCC_July_Map.png', dpi=300, bbox_inches='tight')
plt.show()


```

```python
# === (b) 趋势图：使用 Theil–Sen 中位斜率估计（每年7月 TCC） ===
from scipy.stats import theilslopes

def theil_sen_slope(x, y):
    """自定义 Theil-Sen 斜率：所有 (y_j - y_i)/(x_j - x_i) 中位数"""
    x = np.array(x)
    y = np.array(y)
    slopes = []
    for i in range(len(x)):
        for j in range(i + 1, len(x)):
            if x[j] != x[i]:
                slopes.append((y[j] - y[i]) / (x[j] - x[i]))
    return np.median(slopes) if slopes else np.nan

# 生成趋势图
trend_map = np.full((len(lat), len(lon)), np.nan)

for i in range(len(lat)):
    for j in range(len(lon)):
        y = tcc_sel[:, i, j].values
        valid = np.isfinite(y)
        if np.sum(valid) >= 2:
            x_valid = years[valid]
            y_valid = y[valid]
            trend_map[i, j] = theil_sen_slope(x_valid, y_valid)

# 绘图 (b)
trend_da = xr.DataArray(trend_map, coords={'latitude': lat, 'longitude': lon},
                        dims=['latitude', 'longitude'], name='TCC Trend')
plt.figure(figsize=(10, 8))
trend_da.plot(cmap='RdBu_r', center=0, vmin=-0.002, vmax=0.002,
              cbar_kwargs={'label': 'Trend in TCC (Fraction/year)'})
plt.title('TCC Trend (July, 1950–2024) - Theil–Sen')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.grid(False)
plt.savefig('Greenland_TCC_Trend_TheilSen.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# === (c) 滑动窗口趋势图：允许单年趋势值 ===
min_window_length = 1  # 显式允许 start==end

trend_matrix = np.full((len(years), len(years)), np.nan)

for i, start in enumerate(years):
    for j, end in enumerate(years):
        if end >= start and (end - start + 1) >= min_window_length:
            mask = (years >= start) & (years <= end)
            y = region_mean.sel(year=mask).values
            x = years[mask]
            if np.isfinite(y).sum() > 1:
                slope, _, _, _, _ = linregress(x[np.isfinite(y)], y[np.isfinite(y)])
                trend_matrix[j, i] = slope
            elif end == start:
                trend_matrix[j, i] = 0.0  # 单年设置为 0

# 绘图
plt.figure(figsize=(12, 10))
ax = sns.heatmap(trend_matrix, cmap='RdBu_r', center=0,
                 cbar_kws={'label': 'Trend in TCC (Fraction/year)'},
                 vmin=-0.004, vmax=0.004, square=True)

tick_indices = np.arange(0, len(years), 5)
tick_labels = [str(years[i]) for i in tick_indices]

ax.set_xticks(tick_indices + 0.5)
ax.set_xticklabels(tick_labels, rotation=45, ha='right')

ax.set_yticks(tick_indices + 0.5)
ax.set_yticklabels(tick_labels, rotation=0, va='center')

plt.xlabel('Start Year')
plt.ylabel('End Year')
plt.title('Sliding Window TCC Trend (July, Theil–Sen + Single Year = 0)')
plt.tight_layout()
plt.savefig('Greenland_TCC_SlidingTrend_AllowSingleYear.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import os
import glob
import pandas as pd
import numpy as np

# ———— 用户配置 ————
data_dir = r"F:\TUB\greenlandProject\ERA5LAND\MonthlyFeatures\1995prediction"
years    = list(range(1950, 2025))
months   = ['06', '07', '08']
# ——————————————————

# 1️⃣ 扫描目录下所有 prediction_clip 文件（包括 .csv 或无扩展名）
pattern_all = os.path.join(data_dir, "*prediction_clip*")
all_files   = glob.glob(pattern_all)
csv_files   = [f for f in all_files if f.lower().endswith('.csv')]

print(f"🔍 共发现 {len(all_files)} 个 “prediction_clip” 文件，其中 {len(csv_files)} 个 CSV 文件。")
print("示例文件：", all_files[:5])

# 2️⃣ 针对每个年份和 6–8 月检查是否存在文件
missing = []
found   = {}
for yr in years:
    for m in months:
        # 匹配如 195006prediction_clip.csv 或 195006prediction_clip
        pat = os.path.join(data_dir, f"{yr}{m}prediction_clip*")
        hits = glob.glob(pat)
        if hits:
            found[(yr, m)] = hits[0]   # 取第一个匹配
        else:
            missing.append((yr, m))

print(f"\n❗ 缺失的年份-月份组合共 {len(missing)} 个，示例：", missing[:5])
print(f"✅ 可用的年份-月份组合共 {len(found)} 个。")

# 3️⃣ 读取一个示例 CSV，检查列和数据格式
if csv_files:
    sample = csv_files[0]
    print(f"\n📄 读取示例文件：{os.path.basename(sample)}")
    df_s = pd.read_csv(sample)
    print("Columns:", df_s.columns.tolist())
    print("\n前 5 行：")
    print(df_s.head())
    print("\n数据类型：")
    print(df_s.dtypes)
    print("\n缺失值统计：")
    print(df_s.isna().sum())
    # 确认必备列
    for col in ['latitude', 'longitude', 'albedo']:
        if col not in df_s.columns:
            raise RuntimeError(f"❌ 样本文件缺少必要列：{col}")
else:
    raise RuntimeError("❌ 未发现任何以 .csv 结尾的 prediction_clip 文件，请检查路径或扩展名。")

# 4️⃣ 构建夏季均值时序数据结构，检测能否顺利执行
print("\n📊 开始为每个网格点构建 1950–2024 年夏季 (6–8 月) 均值序列…")
albedo_series = {}
for yr in years:
    dfs = []
    for m in months:
        key = (yr, m)
        if key in found:
            fn = found[key]
            dfm = pd.read_csv(fn)[['latitude','longitude','albedo']]
            dfs.append(dfm)
    if len(dfs) == 3:
        # 三个月文件都齐
        df_sum = pd.concat(dfs).groupby(['latitude','longitude'], as_index=False).mean()
        for _, row in df_sum.iterrows():
            pt = (row.latitude, row.longitude)
            albedo_series.setdefault(pt, []).append(row.albedo)
    else:
        print(f"⚠️ 年份 {yr} 缺少 {3 - len(dfs)} 个文件，跳过该年")

# 5️⃣ 转为 DataFrame 并初步检查
grid_pts = list(albedo_series.keys())
data_mat = [
    albedo_series[k] if len(albedo_series[k]) == len(years)
    else [np.nan] * len(years)
    for k in grid_pts
]
albedo_df = pd.DataFrame(
    data_mat,
    index=pd.MultiIndex.from_tuples(grid_pts, names=['lat','lon']),
    columns=years
)

print("\n✅ albedo_df 构建完毕：")
print(" - 形状 (points × years):", albedo_df.shape)
print(" - 索引示例 (前 5):", list(albedo_df.index[:5]))
print(" - 数据预览 (前 5 × 前 5):")
print(albedo_df.iloc[:5, :5])
print(" - 每年非 NaN 点数：")
print(albedo_df.notna().sum(axis=0).head())
print(" - 每点完整年份计数分布：")
print(albedo_df.notna().sum(axis=1).value_counts().head())

# 如果上述检查无误，就可以放心运行脚本 2/3 中的趋势与滑动相关性计算
```

```python
# A1: 计算夏季均值并保存
import os
import glob
import numpy as np
import pandas as pd
from tqdm import tqdm

# ———— 用户配置 ————
data_dir = r"F:\TUB\greenlandProject\ERA5LAND\MonthlyFeatures\1995prediction"
years    = list(range(1950, 2025))  # 1950–2024
months   = ['06', '07', '08']
# ————————————————————

print("1️⃣ 开始读取并合并每年 6–8 月的 Albedo 数据…")
albedo_series = {}  # {(lat,lon): [albedo1950, albedo1951, ...]}

for year in tqdm(years, desc="读取年份"):
    dfs = []
    for m in months:
        fn_list = glob.glob(os.path.join(data_dir, f"{year}{m}prediction_clip*"))
        if not fn_list:
            continue
        dfm = pd.read_csv(fn_list[0])[['latitude','longitude','albedo']]
        dfs.append(dfm)
    if len(dfs) != 3:
        print(f"⚠️ 年份 {year} 文件不全 ({len(dfs)}/3)，跳过该年")
        continue
    df_sum = pd.concat(dfs).groupby(['latitude','longitude'], as_index=False).mean()
    for _, row in df_sum.iterrows():
        pt = (row.latitude, row.longitude)
        albedo_series.setdefault(pt, []).append(row.albedo)

print("✅ 原始序列读取完毕，共网格点：", len(albedo_series))

print("2️⃣ 构建 DataFrame…")
grid_pts = list(albedo_series.keys())
data_mat = [albedo_series[k] for k in grid_pts]
albedo_df = pd.DataFrame(
    data_mat,
    index=pd.MultiIndex.from_tuples(grid_pts, names=['lat','lon']),
    columns=years
)
print("✅ DataFrame 构建完毕，形状：", albedo_df.shape)

print("3️⃣ 计算夏季均值时序…")
mean_map = albedo_df.mean(axis=1)
mean_map.name = 'mean'
print("✅ 夏季均值计算完毕。")

# 保存到 CSV
out_csv = os.path.join(data_dir, "mean_summer_albedo.csv")
print("4️⃣ 保存 Mean Summer Albedo 到：", out_csv)
mean_map.reset_index().to_csv(out_csv, index=False)
print("✅ 已保存 Mean Summer Albedo。")
```

```python
# B1: 计算 Theil–Sen 趋势并保存
import os
import glob
import numpy as np
import pandas as pd
from scipy.stats import theilslopes
from tqdm import tqdm

# ———— 用户配置 ————
data_dir = r"F:\TUB\greenlandProject\ERA5LAND\MonthlyFeatures\1995prediction"
years    = list(range(1950, 2025))
months   = ['06', '07', '08']
# ————————————————————

print("1️⃣ 开始读取并合并每年 6–8 月的 Albedo 数据…")
albedo_series = {}
for year in tqdm(years, desc="读取年份"):
    dfs = []
    for m in months:
        fn_list = glob.glob(os.path.join(data_dir, f"{year}{m}prediction_clip*"))
        if not fn_list:
            continue
        dfm = pd.read_csv(fn_list[0])[['latitude','longitude','albedo']]
        dfs.append(dfm)
    if len(dfs) != 3:
        print(f"⚠️ 年份 {year} 文件不全 ({len(dfs)}/3)，跳过")
        continue
    df_sum = pd.concat(dfs).groupby(['latitude','longitude'], as_index=False).mean()
    for _, row in df_sum.iterrows():
        pt = (row.latitude, row.longitude)
        albedo_series.setdefault(pt, []).append(row.albedo)
print("✅ 原始序列读取完毕，共网格点：", len(albedo_series))

print("2️⃣ 构建 DataFrame…")
grid_pts = list(albedo_series.keys())
data_mat = [albedo_series[k] for k in grid_pts]
albedo_df = pd.DataFrame(
    data_mat,
    index=pd.MultiIndex.from_tuples(grid_pts, names=['lat','lon']),
    columns=years
)
print("✅ DataFrame 构建完毕，形状：", albedo_df.shape)

print("3️⃣ 计算 Theil–Sen 趋势…")
def calc_theilsen(arr):
    if np.isnan(arr).any():
        return np.nan
    slope, _, _, _ = theilslopes(arr, np.arange(len(arr)))
    return slope

trend_map = albedo_df.apply(calc_theilsen, axis=1)
trend_map.name = 'trend'
print("✅ Theil–Sen 趋势计算完毕。")

# 保存到 CSV
out_csv = os.path.join(data_dir, "theil_sen_trend.csv")
print("4️⃣ 保存 Theil–Sen 趋势到：", out_csv)
trend_map.reset_index().to_csv(out_csv, index=False)
print("✅ 已保存 Theil–Sen 趋势。")
```

```python
# B2: 读取 Theil–Sen 趋势并绘图 - 改进版（非对称拉伸）
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import TwoSlopeNorm

# ———— 用户配置 ————
data_dir = r"F:\TUB\greenlandProject\ERA5LAND\MonthlyFeatures\1995prediction"
# ————————————————————

csv_file = os.path.join(data_dir, "theil_sen_trend.csv")
print("1️⃣ 读取文件：", csv_file)
dft = pd.read_csv(csv_file)  # 列：[lat, lon, trend]

print("2️⃣ 构造透视表并绘图…")
pivot = dft.pivot(index='lat', columns='lon', values='trend')

# 设定我们需要的拉伸范围
trend_lower = -0.003  # 最小值映射到蓝端
trend_upper = +0.001  # 最大值映射到红端

# 创建 TwoSlopeNorm：保证 0 为中心点
norm = TwoSlopeNorm(vmin=trend_lower,
                    vcenter=0.0,
                    vmax=trend_upper)

# 创建图形
fig, ax = plt.subplots(figsize=(10, 6))

im = ax.imshow(
    pivot.values[::-1],
    cmap='RdBu_r',    # 红色 = 正趋势，蓝色 = 负趋势
    norm=norm,        # 使用自定义的中点归一化
    extent=[
        pivot.columns.min(), pivot.columns.max(),
        pivot.index.min(),   pivot.index.max()
    ],
    aspect='auto'
)

# 设置标题和标签
ax.set_title('Greenland_albedo_Trend_Analysis', fontsize=14, pad=20)
ax.set_xlabel('Longitude', fontsize=12)
ax.set_ylabel('Latitude', fontsize=12)
ax.tick_params(axis='both', which='major', labelsize=10)

# 颜色条
cbar = plt.colorbar(im, ax=ax, shrink=1.0, pad=0.02, aspect=20)
cbar.set_label('Trend (unit/year)', fontsize=12)
cbar.ax.tick_params(labelsize=10)
# 可选：自定义刻度，更清晰地显示 -0.003, 0, +0.001
cbar.set_ticks([trend_lower, 0.0, trend_upper])

plt.tight_layout()

# 保存并展示
save_path = os.path.join(data_dir, 'Greenland_albedo_Trend_Analysis.png')
plt.savefig(save_path, dpi=300, bbox_inches='tight')
print(f"图片已保存至: {save_path}")
plt.show()

# 打印统计
trend_min = pivot.values.min()
trend_max = pivot.values.max()
print("3️⃣ 数据统计：")
print(f"   原始趋势范围: {trend_min:.6f} – {trend_max:.6f}")
print(f"   强制拉伸区间: {trend_lower:.6f} – {trend_upper:.6f}")
print(f"   平均趋势: {pivot.values.mean():.6f}")
print(f"   数据形状: {pivot.shape}")
print("🎨 Theil–Sen Trend 绘制完成。")
```

```python
# C1: 计算滑动窗口 Theil–Sen 趋势矩阵并保存
import os, glob
import numpy as np
import pandas as pd
from scipy.stats import theilslopes
from tqdm import tqdm

# ———— 用户配置 ————
data_dir = r"F:\TUB\greenlandProject\ERA5LAND\MonthlyFeatures\1995prediction"
years    = np.array(list(range(1950, 2025)))  # 75 年
months   = ['06','07','08']
out_csv  = os.path.join(data_dir, "sliding_window_trend.csv")
# ————————————————————

print("1️⃣ 计算每年全格陵兰夏季（6–8 月）平均反照率…")
yearly_mean = []
for yr in tqdm(years, desc="年度"):
    # 合并该年 6、7、8 月
    dfs = []
    for m in months:
        fl = glob.glob(os.path.join(data_dir, f"{yr}{m}prediction_clip*"))
        if fl:
            dfs.append(pd.read_csv(fl[0])['albedo'])
    if len(dfs) != 3:
        print(f"⚠️ 年份 {yr} 缺少文件，跳过")
        yearly_mean.append(np.nan)
        continue
    # 每月取全网格平均后再平均三月
    monthly_means = [df.mean() for df in dfs]
    yearly_mean.append(np.mean(monthly_means))

yearly_mean = np.array(yearly_mean)
print("✅ 年度时序计算完毕。")

print("2️⃣ 构建滑动窗口趋势矩阵…")
n = len(years)
# 初始化 DataFrame 全 NaN
trend_mat = pd.DataFrame(
    np.nan, index=years, columns=years, dtype=float
)

# 对每对(start,end)，end>start 计算 Theil–Sen 斜率
for i, start in enumerate(tqdm(years[:-1], desc="滑动起点")):
    for j, end in enumerate(years[i+1:], start=i+1):
        arr = yearly_mean[i:j+1]
        if np.isnan(arr).any(): 
            continue
        slope, _, _, _ = theilslopes(arr, np.arange(len(arr)))
        trend_mat.at[start, end] = slope

print("✅ 滑动窗口趋势矩阵构建完毕，形状：", trend_mat.shape)

print("3️⃣ 保存到 CSV：", out_csv)
trend_mat.to_csv(out_csv, float_format="%.6f")
print("✅ 已保存 Sliding Window Trend 矩阵。")
```

```python
C2显示有问题，参考这个改一下print("1️⃣ 读取矩阵 CSV：", csv_file)
# 第一列 (索引) 是 Start Year，第一行 (列名) 是 End Year。
trend_mat = pd.read_csv(csv_file, index_col=0)

# 获取起始年和结束年列表
start_years = trend_mat.index.astype(int).values
end_years   = trend_mat.columns.astype(int).values

print("2️⃣ 计算色标范围…")
data_vals = np.ma.masked_invalid(trend_mat.values).compressed()
if data_vals.size > 0:
    p2  = np.percentile(data_vals, 2)
    p98 = np.percentile(data_vals, 98)
    M   = max(abs(p2), abs(p98))
    vmin_final, vmax_final = -M, M
else:
    print("⚠️ 无有效数据，使用默认色标")
    vmin_final, vmax_final = -0.01, 0.01

print("3️⃣ 绘图 & 翻转 Y 轴到右上角…")
fig, ax = plt.subplots(figsize=(8, 8))

# 掩盖 NaN
data_for_plot = np.ma.masked_invalid(trend_mat.values)

im = ax.imshow(
    data_for_plot,
    cmap='bwr',
    vmin=vmin_final, vmax=vmax_final,
    extent=[
        end_years.min() - 0.5, end_years.max() + 0.5,
        start_years.min() - 0.5, start_years.max() + 0.5
    ],
    origin='lower',   # 保持 'lower'
    aspect='auto'
)

# —— 关键：把 Y 轴倒过来 —— 
ax.invert_yaxis()

# 坐标和标签
ax.set_xlabel('End Year', fontsize=12)
ax.set_ylabel('Start Year', fontsize=12)
ax.set_title('Greenland Summer Albedo Sliding Trend', fontsize=14)

# 颜色条
cbar = plt.colorbar(im, ax=ax, shrink=0.8)
cbar.set_label('Trend (Albedo / year)', fontsize=10)

# 刻度
num_ticks = 10
xticks = np.linspace(end_years.min(), end_years.max(), num_ticks, dtype=int)
yticks = np.linspace(start_years.min(), start_years.max(), num_ticks, dtype=int)
ax.set_xticks(xticks)
ax.set_yticks(yticks)
ax.tick_params(axis='both', which='major', labelsize=10)

# 网格
ax.grid(which='major', color='grey', linestyle=':', linewidth=0.5)

plt.tight_layout()

# 保存并展示
save_path = os.path.join(data_dir, 'Greenland_SummerAlbedo_Sliding_Trend.png')
plt.savefig(save_path, dpi=300, bbox_inches='tight')
plt.show()
print("🎨 热力图已保存至：", save_path)
```

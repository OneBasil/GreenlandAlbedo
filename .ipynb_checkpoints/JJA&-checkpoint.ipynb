```python
# -*- coding: utf-8 -*-
"""
Step 2 — 月距平(1981–2010) → 季节异常（JJA/AMJJAS）年序列 + Z
- 数据仍取区域平均（cos(lat) 加权）
- 输出CSV：..._region_anom_1981-2010.csv
"""

import numpy as np
import pandas as pd
import xarray as xr

nc_path  = r"E:\TUB\greenland\1995bu\Greenland_albedo1950-2024.nc"
var_name = "albedo"
months_jja    = [6,7,8]
months_amjjas = [4,5,6,7,8,9]
base_start, base_end = 1981, 2010  # 统一和你的TCC等口径

def area_weighted_mean(da, lat_name="lat"):
    w = np.cos(np.deg2rad(da[lat_name]))
    w2 = xr.broadcast(w, da.isel({lat_name: 0}))[0]
    return da.weighted(w2).mean(dim=("lat","lon"))

with xr.open_dataset(nc_path) as ds:
    da = ds[var_name]
    # 逐月区域平均
    ts_month = area_weighted_mean(da).to_series()
    ts_month.index = pd.to_datetime(ds["time"].values)
    ts_month.name = "albedo"

    # 基准期掩膜
    base_mask = (ts_month.index.year >= base_start) & (ts_month.index.year <= base_end)
    ts_base = ts_month[base_mask]
    if ts_base.empty:
        raise RuntimeError("基准期1981–2010在本文件中无数据，请检查时间范围。")

    # —— 月气候态（仅对存在的月份做）——
    clim_monthly = ts_base.groupby(ts_base.index.month).mean()
    # 对现有序列做“逐月距平”
    anom_month = ts_month - ts_month.index.month.map(clim_monthly)

    # —— 聚合为季节异常：JJA与AMJJAS（仅完整年份）——
    def seasonal_yearly_anom(anom_series, months):
        sub = anom_series[anom_series.index.month.isin(months)]
        df = sub.groupby(sub.index.year).agg(["mean","count"])
        df.columns = ["anom","n"]
        df = df[df["n"] == len(months)]
        return df["anom"]

    jja_anom    = seasonal_yearly_anom(anom_month, months_jja)
    amjjas_anom = seasonal_yearly_anom(anom_month, months_amjjas)

    # —— 用基准期年际σ做Z分数 ——（各自季节口径计算σ）
    def zscore(series, base_years):
        base_vals = series[(series.index >= base_years[0]) & (series.index <= base_years[1])]
        sigma = float(base_vals.std(ddof=1))
        if sigma == 0 or np.isnan(sigma):
            z = series * np.nan
        else:
            z = series / sigma
        return z, sigma

    jja_z,    jja_sigma    = zscore(jja_anom,    (base_start, base_end))
    amjjas_z, amjjas_sigma = zscore(amjjas_anom, (base_start, base_end))

    # —— 汇总与导出 ——
    out = pd.DataFrame({
        "albedo_JJA_anom": jja_anom,
        "albedo_JJA_z":    jja_z,
        "albedo_AMJJAS_anom": amjjas_anom,
        "albedo_AMJJAS_z":    amjjas_z
    })
    out.index.name = "year"
    out_path = nc_path.replace(".nc", "_region_anom_1981-2010.csv")
    out.to_csv(out_path, float_format="%.6f")

    # —— 精简提示 —— 
    def flag_years(zs, thr=1.0):
        return zs.index[zs.abs() > thr].tolist()

    print(f"基准期: {base_start}–{base_end}")
    print(f"σ(JJA)={jja_sigma:.4g},  σ(AMJJAS)={amjjas_sigma:.4g}")
    print("JJA |Z|>1 年份：", flag_years(jja_z, 1.0))
    print("AMJJAS |Z|>1 年份：", flag_years(amjjas_z, 1.0))
    print(f"✅ 已导出：{out_path}")
```

```python
# -*- coding: utf-8 -*-
"""
Gridwise MK+Sen (Hamed–Rao) with FDR → CSV + 常见图
- 输入: Greenland_albedo1950-2024.nc (time,lat,lon), 月份只含4-9
- 基准期: 1981–2010 (逐月气候态)
- 季节: JJA(6-8), AMJJAS(4-9)
- 时间窗: 1950–2024, 1979–2024, 1995–2024
- 输出: 每(季节×时间窗)一份CSV + PNG地图 + PNG直方图；附区域序列图
- 低内存：lat分块循环，打印进度；任何异常会报块号并继续/退出（可选）
"""

import os, sys, math, traceback
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt

# ========== 用户配置 ==========
nc_path  = r"E:\TUB\greenland\1995bu\Greenland_albedo1950-2024.nc"
var_name = "albedo"
base_start, base_end = 1981, 2010
seasons = {"JJA":[6,7,8], "AMJJAS":[4,5,6,7,8,9]}
time_windows = [("1950-2024", 1950, 2024),
                ("1979-2024", 1979, 2024),
                ("1995-2024", 1995, 2024)]
min_years = 25       # 每格点最少年数
lat_block = 40       # 每次处理多少条纬线；越小越省内存但更慢
scatter_stride = 3   # 地图上黑点稀疏绘制步长(>=1)，适当减小以提速
dpi = 180            # 出图清晰度
continue_on_error = False  # 某块失败是否继续下一个块
# ============================

try:
    import pymannkendall as mk
except ImportError:
    print("请先安装: pip install pymannkendall"); raise

# ---- 工具函数 ----
def log(msg): 
    print(msg, flush=True)

def bh_fdr(pvals, alpha=0.05):
    """Benjamini–Hochberg FDR，返回显著性布尔数组(同形状)。"""
    p = pvals.copy()
    finite_mask = np.isfinite(p)
    m = finite_mask.sum()
    sig = np.zeros_like(p, dtype=bool)
    if m == 0: 
        return sig
    p_sorted_idx = np.argsort(p[finite_mask], kind="mergesort")
    p_sorted = p[finite_mask][p_sorted_idx]
    ranks = np.arange(1, m+1)
    thresh = alpha * ranks / m
    k = np.where(p_sorted <= thresh)[0]
    cutoff = p_sorted[k.max()] if k.size else -np.inf
    sig[finite_mask] = p[finite_mask] <= cutoff
    return sig

def mk_hr_on_series(y):
    """单格点年序列→MK(Hamed–Rao)+Sen。返回 slope/decade, tau, p, n, z"""
    y = np.asarray(y, dtype=float)
    mask = np.isfinite(y)
    yy = y[mask]
    if yy.size < min_years:
        return np.nan, np.nan, np.nan, float(yy.size), np.nan
    res = mk.hamed_rao_modification_test(yy.tolist())
    slope_decade = float(res.slope) * 10.0
    return slope_decade, float(res.Tau), float(res.p), float(yy.size), float(getattr(res, "z", np.nan))

def seasonal_year_anom_block(da_block, months, base_start, base_end):
    """
    分块数据(时间,lat,lon) → 逐月距平(1981-2010) → 季节年异常(year,lat,lon)
    仅保留完整年（len(months)个月都在）
    """
    base = da_block.sel(time=slice(f"{base_start}-01-01", f"{base_end}-12-31"))
    clim = base.groupby("time.month").mean("time")
    anom = da_block.groupby("time.month") - clim
    # 仅目标月份
    sub = anom.where(anom["time"].dt.month.isin(months), drop=False)
    yr_mean  = sub.groupby("time.year").mean("time")
    yr_count = da_block.where(da_block["time"].dt.month.isin(months)).groupby("time.year").count("time")
    complete = (yr_count == len(months))
    yr_mean  = yr_mean.where(complete)
    yr_mean  = yr_mean.rename({"year":"year"})
    return yr_mean  # (year, lat, lon)

def grid_mk_block(yr_da):
    """对(year,lat,lon)块做MK，返回 slope,tau,p,n,z (lat,lon)。"""
    slope = np.empty(yr_da.shape[1:], dtype=np.float32); slope[:] = np.nan
    tau   = np.empty_like(slope); tau[:] = np.nan
    pval  = np.empty_like(slope); pval[:] = np.nan
    nn    = np.empty_like(slope); nn[:] = np.nan
    zz    = np.empty_like(slope); zz[:] = np.nan

    years = yr_da["year"].values
    # 遍历每个网格列（向量化也行，但分块内循环更稳）
    latN, lonN = slope.shape
    vals = yr_da.values  # shape (year, lat, lon)
    for i in range(latN):
        Y = vals[:, i, :]  # (year, lon)
        # 对每个 lon 做MK
        for j in range(lonN):
            sd, tt, pp, nnv, zzv = mk_hr_on_series(Y[:, j])
            slope[i, j] = sd; tau[i, j] = tt; pval[i, j] = pp; nn[i, j] = nnv; zz[i, j] = zzv
    return slope, tau, pval, nn, zz

def safe_mkdir(path):
    os.makedirs(path, exist_ok=True)

# ---- 主流程 ----
log("打开数据...")
ds = xr.open_dataset(nc_path, decode_times=True)
da = ds[var_name]
lat = ds["lat"].values
lon = ds["lon"].values
ntime, nlat, nlon = da.shape
log(f"数据维度: time={ntime}, lat={nlat}, lon={nlon}")

out_dir = os.path.dirname(nc_path)
img_dir = os.path.join(out_dir, "fig_grid_mk")
csv_dir = os.path.join(out_dir, "csv_grid_mk")
safe_mkdir(img_dir); safe_mkdir(csv_dir)

# 区域经纬度2D网格（用于CSV展开）
Lon2D, Lat2D = np.meshgrid(lon, lat)

def run_one_season(season_name, months):
    log(f"\n==== 处理季节: {season_name} 月份{months} ====")
    # 预先计算季节年异常（逐块），并缓存为按块的列表，避免重复做两遍窗口
    # 为了低内存，这里不把所有年异常拼成一个大array，而是：在每个时间窗时再筛选年份
    # 但为了做多窗，我们需要先在一次块运算内得到完整(year,lat_block,lon)数据
    # 我们将按lat块把季节年异常写入一个临时list（内存装得下，因为每块年×lat_block×lon）
    blocks = []  # list of dict: {"lat_slice":(i0,i1), "year": years, "anom": ndarray(year, b, nlon)}
    years_union = None

    # 分块构建季节年异常
    for i0 in range(0, nlat, lat_block):
        i1 = min(i0 + lat_block, nlat)
        log(f"[{season_name}] 距平→年异常  构建块 lat[{i0}:{i1}) ...")
        try:
            block = da.isel(lat=slice(i0, i1))
            yr_da = seasonal_year_anom_block(block, months, base_start, base_end)  # (year, b, nlon)
            years = yr_da["year"].values
            if years_union is None:
                years_union = years
            else:
                # 对齐年份；保守起见，取两者交集（通常一致）
                years_union = np.intersect1d(years_union, years)
            blocks.append({"lat_slice":(i0,i1), "year": years, "anom": yr_da.values})
        except Exception as e:
            log(f"❌ 构建块失败 lat[{i0}:{i1}): {e}")
            traceback.print_exc()
            if not continue_on_error: 
                raise

    # 对每个时间窗做MK，收集到全域数组→FDR→CSV+图
    for tag, y0, y1 in time_windows:
        log(f"\n[{season_name}] 时间窗 {tag} ({y0}-{y1}) → MK+Sen+FDR ...")
        slope_full = np.full((nlat, nlon), np.nan, dtype=np.float32)
        tau_full   = np.full_like(slope_full, np.nan)
        p_full     = np.full_like(slope_full, np.nan)
        n_full     = np.full_like(slope_full, np.nan)
        z_full     = np.full_like(slope_full, np.nan)

        # 各块MK
        for b, info in enumerate(blocks):
            i0, i1 = info["lat_slice"]; years = info["year"]; anom = info["anom"]  # (year, b, nlon)
            # 找到窗内年份索引
            yr_mask = (years >= y0) & (years <= y1)
            if not np.any(yr_mask):
                log(f"  ⚠️ 块 lat[{i0}:{i1}) 在 {tag} 窗口无年份，跳过")
                continue
            anom_win = anom[yr_mask, :, :]
            log(f"  [{season_name}/{tag}] MK 块 lat[{i0}:{i1}) 年数={anom_win.shape[0]}")
            try:
                slope, tau, pval, nn, zz = grid_mk_block(xr.DataArray(anom_win, dims=["year","lat","lon"]))
                slope_full[i0:i1, :] = slope
                tau_full[i0:i1, :]   = tau
                p_full[i0:i1, :]     = pval
                n_full[i0:i1, :]     = nn
                z_full[i0:i1, :]     = zz
            except Exception as e:
                log(f"  ❌ MK失败 lat[{i0}:{i1})：{e}")
                traceback.print_exc()
                if not continue_on_error: 
                    raise

        # FDR across space
        log(f"[{season_name}/{tag}] FDR 计算 ...")
        signif = bh_fdr(p_full, alpha=0.05)

        # → CSV（长表）
        df = pd.DataFrame({
            "lat":  Lat2D.ravel(),
            "lon":  Lon2D.ravel(),
            "slope_decade": slope_full.ravel(),
            "tau":  tau_full.ravel(),
            "p":    p_full.ravel(),
            "z":    z_full.ravel(),
            "n":    n_full.ravel(),
            "signif_fdr": signif.ravel().astype(int)
        })
        csv_path = os.path.join(csv_dir, f"albedo_MK_{season_name}_{tag}_grid.csv")
        df.to_csv(csv_path, index=False, float_format="%.6g")
        log(f"[{season_name}/{tag}] ✅ CSV: {csv_path}")

        # ——— 常见图 1：趋势地图（slope/decade + FDR黑点） ———
        try:
            fig, ax = plt.subplots(figsize=(9, 6), dpi=dpi)
            im = ax.imshow(slope_full, origin="lower",
                           extent=[lon.min(), lon.max(), lat.min(), lat.max()],
                           cmap="RdBu_r", vmin=-0.02, vmax=0.02, interpolation="bilinear")
            ax.set_title(f"Albedo Sen slope per decade ({season_name}, {tag})\nHamed–Rao MK; black dots: FDR q<0.05")
            ax.set_xlabel("Lon"); ax.set_ylabel("Lat")
            cb = plt.colorbar(im, ax=ax, shrink=0.8, label="albedo / decade")
            # 黑点（可稀疏）
            yy, xx = np.where(signif)
            if scatter_stride > 1:
                yy = yy[::scatter_stride]; xx = xx[::scatter_stride]
            ax.scatter(lon[xx], lat[yy], s=1, c="k")
            fig.tight_layout()
            fig_path = os.path.join(img_dir, f"map_slope_{season_name}_{tag}.png")
            fig.savefig(fig_path); plt.close(fig)
            log(f"[{season_name}/{tag}] 🗺️ 地图: {fig_path}")
        except Exception as e:
            log(f"  ⚠️ 地图绘制失败（可忽略）：{e}")

        # ——— 常见图 2：slope直方图 ———
        try:
            vals = slope_full[np.isfinite(slope_full)]
            fig, ax = plt.subplots(figsize=(7,5), dpi=dpi)
            ax.hist(vals, bins=60)
            ax.set_title(f"Slope histogram ({season_name}, {tag})")
            ax.set_xlabel("albedo / decade"); ax.set_ylabel("count")
            fig.tight_layout()
            fig_path = os.path.join(img_dir, f"hist_slope_{season_name}_{tag}.png")
            fig.savefig(fig_path); plt.close(fig)
            log(f"[{season_name}/{tag}] 📊 直方图: {fig_path}")
        except Exception as e:
            log(f"  ⚠️ 直方图绘制失败（可忽略）：{e}")

# 先画区域均值时间序列（异常，基准1981–2010），并叠加不同时间窗的线性拟合
def area_weighted_mean(da, lat_name="lat"):
    w = np.cos(np.deg2rad(da[lat_name])); w = w / w.mean()
    w2 = xr.broadcast(w, da.isel({lat_name:0}))[0]
    return da.weighted(w2).mean(dim=("lat","lon"))

def region_series_and_plots():
    log("\n==== 区域均值时间序列与分时间窗线性拟合 ====")
    ts_month = area_weighted_mean(da).to_series()
    ts_month.index = pd.to_datetime(ds["time"].values)
    # 月气候态（1981–2010）
    base_mask = (ts_month.index.year >= base_start) & (ts_month.index.year <= base_end)
    clim_month = ts_month[base_mask].groupby(ts_month[base_mask].index.month).mean()
    anom_month = ts_month - ts_month.index.month.map(clim_month)

    for season_name, months in seasons.items():
        sub = anom_month[anom_month.index.month.isin(months)]
        df = sub.groupby(sub.index.year).agg(["mean","count"])
        df.columns = ["anom","n"]; df = df[df["n"]==len(months)]
        ser = df["anom"]; ser.index.name = "year"
        # 保存CSV
        csv_path = os.path.join(csv_dir, f"region_anom_{season_name}_1981-2010.csv")
        ser.to_csv(csv_path, header=[f"albedo_{season_name}_anom"], float_format="%.6f")
        log(f"[{season_name}] ✅ 区域年异常CSV: {csv_path}")

        # 画时间序列 + 线性拟合(各时间窗)
        try:
            fig, ax = plt.subplots(figsize=(9,5), dpi=dpi)
            ax.plot(ser.index, ser.values, lw=1.2)
            ax.axhline(0, ls="--", lw=0.8, color="k")
            ax.set_title(f"Regional albedo anomaly series ({season_name}, base 1981–2010)")
            ax.set_xlabel("Year"); ax.set_ylabel("albedo anomaly")
            # 分时间窗拟合
            for tag,y0,y1 in time_windows:
                m = (ser.index>=y0)&(ser.index<=y1)
                if m.sum()>=min_years:
                    x = ser.index[m].values.astype(float)
                    y = ser.values[m]
                    # 用简单OLS做趋势线展示（MK结果在格点层面已给）
                    A = np.vstack([x, np.ones_like(x)]).T
                    slope, intercept = np.linalg.lstsq(A, y, rcond=None)[0]
                    ax.plot(ser.index[m], slope*ser.index[m].values.astype(float)+intercept, lw=1.5, label=f"{tag}")
            ax.legend(frameon=False)
            fig.tight_layout()
            fig_path = os.path.join(img_dir, f"region_series_{season_name}.png")
            fig.savefig(fig_path); plt.close(fig)
            log(f"[{season_name}] 🧵 区域序列图: {fig_path}")
        except Exception as e:
            log(f"  ⚠️ 区域序列绘制失败（可忽略）：{e}")

# ===== 执行 =====
for sn, months in seasons.items():
    run_one_season(sn, months)

region_series_and_plots()
log("\n🎉 全部完成。CSV在: " + csv_dir + " ；图在: " + img_dir)
```

```python
# -*- coding: utf-8 -*-
import os
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
from matplotlib.colors import TwoSlopeNorm
from matplotlib.ticker import FuncFormatter, FormatStrFormatter
import matplotlib.gridspec as gridspec

# === 路径与参数 ===
nc_path   = r"E:\TUB\greenland\1995bu\Greenland_albedo1950-2024.nc"
var_name  = "albedo"
base_rng  = (1981, 2010)       # 基准期
season    = [6, 7, 8]          # JJA
years     = list(range(2011, 2025))  # 2011–2024
cols, rows = 5, 3               # 每行5张，共3行
save_png  = os.path.join(os.path.dirname(nc_path), "fig_JJA_base_anoms_1981-2010_2011-2024.png")

# === 辅助函数 ===
def panel_size(lats, lons, base_panel_w=3.0, scale=1.18):
    """
    依据经纬度纵横比估算单个面板的宽高，避免图像拉伸变形。
    返回 (单面板宽, 单面板高)
    """
    lon_span = float(np.nanmax(lons) - np.nanmin(lons))
    lat_span = float(np.nanmax(lats) - np.nanmin(lats))
    panel_h = base_panel_w * (lat_span / max(lon_span, 1e-6)) * scale
    return base_panel_w, panel_h

def format_axis(ax, show_xlabel=False, show_ylabel=False):
    for s in ["top", "right"]:
        ax.spines[s].set_visible(False)
    ax.tick_params(axis="both", which="both", labelsize=8, length=2)
    if show_xlabel:
        ax.set_xlabel("Longitude", fontsize=9)
    else:
        ax.set_xlabel("")
        ax.set_xticklabels([])
    if show_ylabel:
        ax.set_ylabel("Latitude", fontsize=9)
    else:
        ax.set_ylabel("")
        ax.set_yticklabels([])

def robust_limits(arr, pct=(2, 98), symmetric=False):
    vals = np.asarray(arr).ravel()
    vals = vals[np.isfinite(vals)]
    if vals.size == 0:
        return (-1, 1) if symmetric else (0, 1)
    p2, p98 = np.nanpercentile(vals, pct)
    if symmetric:
        M = max(abs(p2), abs(p98))
        return (-M, M)
    return (p2, p98)

# === 读取数据 ===
ds = xr.open_dataset(nc_path, decode_times=True, mask_and_scale=True)
da = ds[var_name]  # (time, lat, lon)
lat = ds["lat"].values
lon = ds["lon"].values

# === 基准期 JJA 均值（先按每年JJA均值，再 across years 均值）===
sub_base = da.sel(time=slice(f"{base_rng[0]}-01-01", f"{base_rng[1]}-12-31"))
sub_base = sub_base.where(sub_base["time"].dt.month.isin(season), drop=False)
base_yearly = sub_base.groupby("time.year").mean("time", skipna=True)
base_map = base_yearly.mean("year", skipna=True)  # (lat, lon)

# === 2011–2024 各年的 JJA 距平 ===
anom_maps = {}
for y in years:
    sub_y = da.sel(time=slice(f"{y}-01-01", f"{y}-12-31"))
    sub_y = sub_y.where(sub_y["time"].dt.month.isin(season), drop=False)
    year_map = sub_y.mean("time", skipna=True)
    anom_maps[y] = (year_map - base_map)

# === 配色范围 ===
# 基准期面板: 稳健范围
base_vmin, base_vmax = robust_limits(base_map.values, pct=(2, 98), symmetric=False)
# 距平统一色标（对称），基于所有距平的2–98分位
all_anoms = np.stack([anom_maps[y].values for y in years], axis=0)
anom_vmin, anom_vmax = robust_limits(all_anoms, pct=(2, 98), symmetric=True)

# 缺测颜色 & 色图
base_cmap = plt.get_cmap("cividis").copy()
anom_cmap = plt.get_cmap("RdBu_r").copy()
base_cmap.set_bad("lightgray")
anom_cmap.set_bad("lightgray")

# === 画布大小（按面板拼接计算，预留边距避免重叠） ===
panel_w, panel_h = panel_size(lat, lon, base_panel_w=3.2, scale=1.18)
fig_w = panel_w * cols + 1.2   # 额外留边，避免重叠
fig_h = panel_h * (1 + rows) + 1.6
fig = plt.figure(figsize=(fig_w, fig_h), dpi=220)

# GridSpec：1 行（基准）+ 3 行（距平）
gs = gridspec.GridSpec(1 + rows, cols,
                       height_ratios=[1.05] + [1.0]*rows,
                       hspace=0.26, wspace=0.08)

# === 顶部：基准期 JJA 均值 ===
ax0 = fig.add_subplot(gs[0, :])
im0 = ax0.imshow(
    base_map.values, origin="lower",
    extent=[lon.min(), lon.max(), lat.min(), lat.max()],
    cmap=base_cmap, vmin=base_vmin, vmax=base_vmax, interpolation="bilinear"
)
# 顶部只显示左右两边的刻度以免拥挤
for s in ["top", "right"]:
    ax0.spines[s].set_visible(False)
ax0.tick_params(axis="both", which="both", labelsize=9, length=3)
ax0.set_xlabel("Longitude", fontsize=10)
ax0.set_ylabel("Latitude", fontsize=10)
ax0.set_title(f"Baseline JJA mean albedo (1981–2010)", fontsize=11, pad=6)

# === 距平面板（3 行 × 5 列；最后 1 个槽位自动留空） ===
axes_anom = []
last_im = None
for idx, y in enumerate(years):
    r = idx // cols   # 0..2
    c = idx % cols
    ax = fig.add_subplot(gs[1 + r, c])
    im = ax.imshow(
        anom_maps[y].values, origin="lower",
        extent=[lon.min(), lon.max(), lat.min(), lat.max()],
        cmap=anom_cmap, norm=TwoSlopeNorm(vmin=anom_vmin, vcenter=0.0, vmax=anom_vmax),
        interpolation="bilinear"
    )
    # 只在最左列显示Y轴标签、最底行显示X轴标签，避免拥挤与重叠
    show_ylabel = (c == 0)
    show_xlabel = (r == (rows - 1))
    format_axis(ax, show_xlabel=show_xlabel, show_ylabel=show_ylabel)
    ax.set_title(f"{y}  JJA anomaly vs 1981–2010", fontsize=9, pad=4)
    axes_anom.append(ax)
    last_im = im

# 若不足 rows*cols 的槽位，隐藏剩余空轴，防止空白边框
total_slots = rows * cols
if len(years) < total_slots:
    for k in range(len(years), total_slots):
        r = k // cols
        c = k % cols
        ax_empty = fig.add_subplot(gs[1 + r, c])
        ax_empty.axis("off")

# === 色条（放在图外，避免与子图重叠） ===
# 顶部：基准色条
cax0 = fig.add_axes([0.08, 0.92, 0.84, 0.018])  # 左、下、宽、高（相对坐标）
cb0 = fig.colorbar(im0, cax=cax0, orientation="horizontal")
cb0.ax.tick_params(labelsize=9, length=3)
cb0.formatter = FormatStrFormatter("%.02f")
cb0.update_ticks()
cb0.set_label("albedo", fontsize=10)

# 底部：距平色条（所有距平共享）
cax1 = fig.add_axes([0.08, 0.05, 0.84, 0.018])
cb1 = fig.colorbar(last_im, cax=cax1, orientation="horizontal")
cb1.ax.tick_params(labelsize=9, length=3)
cb1.formatter = FuncFormatter(lambda x, pos: f"{x:.03f}")
cb1.update_ticks()
cb1.set_label("albedo anomaly", fontsize=10)

# 布局：为上下色条留出空间，确保“所有图不重叠”
fig.tight_layout(rect=[0.06, 0.08, 0.98, 0.90])

# === 保存与信息 ===
fig.savefig(save_png, dpi=220)
plt.show()

print(
    f"✅ Saved: {save_png}\n"
    f"Baseline range ≈ [{base_vmin:.2f}, {base_vmax:.2f}]  |  "
    f"Anomaly (sym) range ≈ [{anom_vmin:.3f}, {anom_vmax:.3f}]\n"
    f"Years: {years[0]}–{years[-1]} (n={len(years)}), layout: {rows}×{cols}, "
    f"empty_slots={rows*cols - len(years)}"
)
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ========== 路径 ==========
dem_path  = r"E:\TUB\greenland\greenland_DEM_01deg.csv"
mk_dir    = r"E:\TUB\greenland\1995bu\csv_grid_mk"
mk_files  = {
    "1950-2024": os.path.join(mk_dir, "albedo_MK_JJA_1950-2024_grid.csv"),
    "1979-2024": os.path.join(mk_dir, "albedo_MK_JJA_1979-2024_grid.csv"),
    "1995-2024": os.path.join(mk_dir, "albedo_MK_JJA_1995-2024_grid.csv"),
}
out_dir   = os.path.join(mk_dir, "plots_with_dem")
os.makedirs(out_dir, exist_ok=True)

# ========== 参数 ==========
value_col           = "slope_decade"        # 底图变量
sig_col             = "signif_fdr"          # 显著性列
elev_col_in_dem     = "altitude_last_valid" # DEM 高程列
elev_min_for_plot   = 0.0                   # 等高线仅画 >=0 m
sig_stride          = 3                     # 显著性点抽稀：0.1°格网取每 0.3°
dpi                 = 300

# 等高线样式
contour_interval    = 250.0                 # 等高线间距：250 m
label_every         = 500.0                 # 每隔多少米加数字标（避免过密；要每条都标就设为 250.0）
contour_color       = "#222222"             # 深色等高线
contour_linewidth   = 0.8                   # 线宽

# ========== 工具函数 ==========
def auto_symmetric_vmin_vmax(vals, p_lo=2, p_hi=98):
    v = np.asarray(vals, dtype=float)
    v = v[np.isfinite(v)]
    if v.size == 0:
        return -1, 1
    p2, p98 = np.percentile(v, [p_lo, p_hi])
    M = max(abs(p2), abs(p98))
    if not np.isfinite(M) or M == 0:
        M = np.nanmax(np.abs(v)) if np.any(np.isfinite(v)) else 1.0
        if not np.isfinite(M) or M == 0: M = 1.0
    return -M, M

def grid_from_points(df, val_col):
    d = df[["lat","lon",val_col]].dropna()
    lats = np.sort(d["lat"].unique())
    lons = np.sort(d["lon"].unique())
    pivot = d.pivot(index="lat", columns="lon", values=val_col).reindex(index=lats, columns=lons)
    Lon2D, Lat2D = np.meshgrid(lons, lats)
    Val2D = pivot.values
    return Lon2D, Lat2D, Val2D

def thin_sig_points(df, stride=3):
    # 0.1° 格网 → stride=3 表示每 0.3° 取一个
    lat_i = np.round(df["lat"].values * 10).astype(int)
    lon_i = np.round(df["lon"].values * 10).astype(int)
    keep = (lat_i % stride == 0) & (lon_i % stride == 0)
    return df[keep]

# ========== 读 DEM ==========
dem = pd.read_csv(dem_path)
dem.columns = [c.strip() for c in dem.columns]
lat_dem = "latitude" if "latitude" in dem.columns else "lat"
lon_dem = "longitude" if "longitude" in dem.columns else "lon"
if elev_col_in_dem not in dem.columns:
    raise ValueError(f"DEM未找到高程列 {elev_col_in_dem}；现有列: {list(dem.columns)}")

dem = dem.rename(columns={lat_dem: "lat", lon_dem: "lon", elev_col_in_dem: "elev"})
dem["lat"] = dem["lat"].round(3)
dem["lon"] = dem["lon"].round(3)
dem = dem[["lat","lon","elev"]]

# ========== 主流程 ==========
for tag, mk_path in mk_files.items():
    print(f"\n== 处理 {tag} ==")
    df = pd.read_csv(mk_path)
    df.columns = [c.strip() for c in df.columns]
    df = df.rename(columns={"latitude":"lat","longitude":"lon"})
    if "lat" not in df.columns or "lon" not in df.columns:
        raise ValueError(f"{os.path.basename(mk_path)} 未找到经纬度列；现有列: {list(df.columns)}")

    # 对齐坐标精度
    df["lat"] = df["lat"].round(3)
    df["lon"] = df["lon"].round(3)

    # 合并 DEM
    merged = df.merge(dem, on=["lat","lon"], how="left")
    merged_csv = os.path.join(out_dir, f"MK_JJA_{tag}_with_dem.csv")
    merged.to_csv(merged_csv, index=False)
    print("已保存合并CSV:", merged_csv)

    # 用全部可用点构网格画底图
    use = merged.copy()
    LON, LAT, Ztrend = grid_from_points(use, value_col)
    vmin, vmax = auto_symmetric_vmin_vmax(use[value_col].values, 2, 98)

    # ==== 绘图 ====
    fig, ax = plt.subplots(figsize=(8.5, 7), dpi=dpi)

    # 趋势底图
    im = ax.pcolormesh(LON, LAT, Ztrend, shading="nearest", cmap="RdBu_r", vmin=vmin, vmax=vmax, zorder=1)

    # 等高线（tricontour，深色+数字标）
    dem_pts = use[(use["elev"].notna()) & (use["elev"] >= elev_min_for_plot)]
    if not dem_pts.empty:
        max_elev = float(np.nanmax(dem_pts["elev"].values))
        if max_elev < contour_interval:
            levels = [0.0]
        else:
            levels = np.arange(0.0, np.ceil(max_elev/contour_interval)*contour_interval + 1e-6, contour_interval)

        try:
            cs = ax.tricontour(
                dem_pts["lon"].values, dem_pts["lat"].values, dem_pts["elev"].values,
                levels=levels, colors=contour_color, linewidths=contour_linewidth, zorder=2
            )
            # 数字标注（默认每 500 m 标一次，避免过密；想每条都标就把 label_every=250.0）
            label_levels = [lv for lv in levels if (lv % label_every == 0)]
            if len(label_levels) == 0:  # 兜底：至少标所有线
                label_levels = levels
            ax.clabel(cs, levels=label_levels, fmt=lambda x: f"{int(round(x))}", fontsize=7, inline=True)
        except Exception as e:
            print("等高线绘制失败(tricontour)：", e)
    else:
        print("无可用于等高线的 DEM 点。")

    # 显著性点（抽稀）
    if sig_col in use.columns:
        sig_pts = use[(use[sig_col]==1) & use[value_col].notna()]
        if not sig_pts.empty:
            sig_pts = thin_sig_points(sig_pts, stride=sig_stride)
            ax.scatter(sig_pts["lon"], sig_pts["lat"], s=5, c="k", alpha=0.25, linewidths=0,
                       label=f"FDR-significant (/{sig_stride} grid)", zorder=3)

    # 轴/色标/标题
    cb = fig.colorbar(im, ax=ax, shrink=0.9, pad=0.02)
    cb.set_label("Albedo trend per decade (Δ/decade)")

    ax.set_title(f"JJA Albedo Trend ({tag}) with DEM Contours")
    ax.set_xlabel("Longitude (°)")
    ax.set_ylabel("Latitude (°)")
    ax.set_xlim(np.nanmin(LON), np.nanmax(LON))
    ax.set_ylim(np.nanmin(LAT), np.nanmax(LAT))
    ax.set_aspect("equal", adjustable="box")

    if sig_col in use.columns and 'sig_pts' in locals() and not sig_pts.empty:
        ax.legend(loc="lower left", frameon=True, fontsize=8)

    fig.tight_layout()
    png_path = os.path.join(out_dir, f"MK_JJA_{tag}_trend_dem_contours.png")
    fig.savefig(png_path, dpi=dpi)
    plt.close(fig)
    print("已保存图像：", png_path)

print("\n完成 ✅ 去 out_dir 看 PNG 和合并后的 CSV。")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ========= 路径 =========
mk_dir = r"E:\TUB\greenland\1995bu\csv_grid_mk"
mk_files = {
    "1950-2024": os.path.join(mk_dir, "albedo_MK_JJA_1950-2024_grid.csv"),
    "1979-2024": os.path.join(mk_dir, "albedo_MK_JJA_1979-2024_grid.csv"),
    "1995-2024": os.path.join(mk_dir, "albedo_MK_JJA_1995-2024_grid.csv"),
}
out_dir = os.path.join(mk_dir, "plots_with_dem")  # 仍然输出到这个目录
os.makedirs(out_dir, exist_ok=True)

# 两个边界矢量
coast_shp = r"E:\TUB\greenland\Greenland_shp\Gboundry.shp"
icesheet_shp = r"E:\TUB\greenland\Greenland_shp\流域冰盖\格陵兰边界.shp"

# ========= 参数 =========
value_col   = "slope_decade"   # 趋势底图字段
sig_col     = "signif_fdr"     # 显著性字段（1=显著）
dpi         = 300
sig_stride  = 3                # 显著性点抽稀（0.1°格网→每0.3°取一个）

# 边界样式
coast_color, coast_lw = "#000000", 1.0      # 海岸线/外边界：黑色
ice_color, ice_lw     = "#444444", 0.9      # 冰盖边界：深灰

# ========= 工具函数 =========
def auto_symmetric_vmin_vmax(vals, p_lo=2, p_hi=98):
    v = np.asarray(vals, dtype=float)
    v = v[np.isfinite(v)]
    if v.size == 0:
        return -1, 1
    p2, p98 = np.percentile(v, [p_lo, p_hi])
    M = max(abs(p2), abs(p98))
    if not np.isfinite(M) or M == 0:
        M = np.nanmax(np.abs(v)) if np.any(np.isfinite(v)) else 1.0
        if not np.isfinite(M) or M == 0: M = 1.0
    return -M, M

def grid_from_points(df, val_col):
    d = df[["lat","lon",val_col]].dropna()
    lats = np.sort(d["lat"].unique())
    lons = np.sort(d["lon"].unique())
    pivot = d.pivot(index="lat", columns="lon", values=val_col).reindex(index=lats, columns=lons)
    Lon2D, Lat2D = np.meshgrid(lons, lats)
    Val2D = pivot.values
    return Lon2D, Lat2D, Val2D

def thin_sig_points(df, stride=3):
    lat_i = np.round(df["lat"].values * 10).astype(int)
    lon_i = np.round(df["lon"].values * 10).astype(int)
    keep = (lat_i % stride == 0) & (lon_i % stride == 0)
    return df[keep]

def load_boundaries():
    gdf_coast, gdf_ice = None, None
    try:
        import geopandas as gpd
    except Exception as e:
        print("⚠ 未安装 geopandas，无法绘制边界：", e)
        return None, None

    if os.path.exists(coast_shp):
        try:
            gdf_coast = gpd.read_file(coast_shp)
            if gdf_coast.crs is not None:
                gdf_coast = gdf_coast.to_crs(epsg=4326)
        except Exception as e:
            print("⚠ 读取外边界失败：", e)
            gdf_coast = None
    else:
        print("⚠ 未找到外边界文件：", coast_shp)

    if os.path.exists(icesheet_shp):
        try:
            gdf_ice = gpd.read_file(icesheet_shp)
            if gdf_ice.crs is not None:
                gdf_ice = gdf_ice.to_crs(epsg=4326)
        except Exception as e:
            print("⚠ 读取冰盖边界失败：", e)
            gdf_ice = None
    else:
        print("⚠ 未找到冰盖边界文件：", icesheet_shp)

    return gdf_coast, gdf_ice

# ========= 读边界 =========
gdf_coast, gdf_ice = load_boundaries()

# ========= 主流程 =========
for tag, mk_path in mk_files.items():
    print(f"\n== 处理 {tag} ==")
    df = pd.read_csv(mk_path)
    df.columns = [c.strip() for c in df.columns]
    df = df.rename(columns={"latitude":"lat","longitude":"lon"})
    if "lat" not in df.columns or "lon" not in df.columns:
        raise ValueError(f"{os.path.basename(mk_path)} 未找到经纬度列；现有列: {list(df.columns)}")

    df["lat"] = df["lat"].round(3)
    df["lon"] = df["lon"].round(3)

    # pcolormesh 网格
    LON, LAT, Ztrend = grid_from_points(df, value_col)
    vmin, vmax = auto_symmetric_vmin_vmax(df[value_col].values, 2, 98)

    # === 绘图 ===
    fig, ax = plt.subplots(figsize=(8.5, 7), dpi=dpi)
    im = ax.pcolormesh(LON, LAT, Ztrend, shading="nearest", cmap="RdBu_r", vmin=vmin, vmax=vmax, zorder=1)

    # 叠加边界
    if gdf_coast is not None and not gdf_coast.empty:
        try:
            gdf_coast.plot(ax=ax, facecolor="none", edgecolor=coast_color, linewidth=coast_lw, zorder=3)
        except Exception as e:
            print("⚠ 外边界绘制失败：", e)
    if gdf_ice is not None and not gdf_ice.empty:
        try:
            gdf_ice.plot(ax=ax, facecolor="none", edgecolor=ice_color, linewidth=ice_lw, zorder=3)
        except Exception as e:
            print("⚠ 冰盖边界绘制失败：", e)

    # 显著性点（抽稀）
    if sig_col in df.columns:
        sig_pts = df[(df[sig_col]==1) & df[value_col].notna()]
        if not sig_pts.empty:
            sig_pts = thin_sig_points(sig_pts, stride=sig_stride)
            ax.scatter(sig_pts["lon"], sig_pts["lat"], s=5, c="k", alpha=0.25, linewidths=0, zorder=4,
                       label=f"FDR-significant (/{sig_stride} grid)")

    cb = fig.colorbar(im, ax=ax, shrink=0.9, pad=0.02)
    cb.set_label("Albedo trend per decade (Δ/decade)")

    ax.set_title(f"JJA Albedo Trend ({tag}) with Boundaries")
    ax.set_xlabel("Longitude (°)")
    ax.set_ylabel("Latitude (°)")
    ax.set_xlim(np.nanmin(LON), np.nanmax(LON))
    ax.set_ylim(np.nanmin(LAT), np.nanmax(LAT))
    ax.set_aspect("equal", adjustable="box")
    if 'sig_pts' in locals() and not sig_pts.empty:
        ax.legend(loc="lower left", frameon=True, fontsize=8)

    fig.tight_layout()
    png_path = os.path.join(out_dir, f"MK_JJA_{tag}_trend_with_boundaries.png")
    fig.savefig(png_path, dpi=dpi)
    plt.close(fig)
    print("已保存图像：", png_path)

print("\n完成 ✅ 去 out_dir 查看 PNG。")
```

```python
import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ========== 路径 ==========
dem_path   = r"E:\TUB\greenland\greenland_DEM_01deg.csv"  # 仅用于海面掩膜（可关）
mk_dir     = r"E:\TUB\greenland\1995bu\csv_grid_mk"
mk_files   = {
    "1950-2024": os.path.join(mk_dir, "albedo_MK_JJA_1950-2024_grid.csv"),
    "1979-2024": os.path.join(mk_dir, "albedo_MK_JJA_1979-2024_grid.csv"),
    "1995-2024": os.path.join(mk_dir, "albedo_MK_JJA_1995-2024_grid.csv"),
}
# 递归搜 .shp 的根目录
basemap_root = r"E:\TUB\greenland\Greenland_shp\流域冰盖"

out_dir   = os.path.join(mk_dir, "plots_with_dem")
os.makedirs(out_dir, exist_ok=True)

# ========== 参数 ==========
value_col           = "slope_decade"        # 底图变量
sig_col             = "signif_fdr"          # 显著性列
elev_col_in_dem     = "altitude_last_valid" # DEM 高程列
elev_min_for_plot   = 0.0                   # 只显示陆地/冰盖（>=0 m）；想全域就设为 -np.inf
sig_stride          = 3                     # 显著性点抽稀：0.1°格网取每 0.3°
dpi                 = 300

# 底图 shp 样式
edge_colors_cycle   = ["#111111","#444444","#006d77","#7f5539","#1d3557","#2a9d8f","#6a4c93"]
edge_linewidth      = 0.9
edge_alpha          = 0.9

# ========== 工具函数 ==========
def auto_symmetric_vmin_vmax(vals, p_lo=2, p_hi=98):
    v = np.asarray(vals, dtype=float)
    v = v[np.isfinite(v)]
    if v.size == 0:
        return -1, 1
    p2, p98 = np.percentile(v, [p_lo, p_hi])
    M = max(abs(p2), abs(p98))
    if not np.isfinite(M) or M == 0:
        M = np.nanmax(np.abs(v)) if np.any(np.isfinite(v)) else 1.0
        if not np.isfinite(M) or M == 0: M = 1.0
    return -M, M

def grid_from_points(df, val_col):
    d = df[["lat","lon",val_col]].dropna()
    lats = np.sort(d["lat"].unique())
    lons = np.sort(d["lon"].unique())
    pivot = d.pivot(index="lat", columns="lon", values=val_col).reindex(index=lats, columns=lons)
    Lon2D, Lat2D = np.meshgrid(lons, lats)
    Val2D = pivot.values
    return Lon2D, Lat2D, Val2D

def thin_sig_points(df, stride=3):
    # 0.1° 格网 → stride=3 表示每 0.3° 取一个
    lat_i = np.round(df["lat"].values * 10).astype(int)
    lon_i = np.round(df["lon"].values * 10).astype(int)
    keep = (lat_i % stride == 0) & (lon_i % stride == 0)
    return df[keep]

def read_dem_mask(dem_path, elev_col, elev_min):
    dem = pd.read_csv(dem_path)
    dem.columns = [c.strip() for c in dem.columns]
    lat_dem = "latitude" if "latitude" in dem.columns else "lat"
    lon_dem = "longitude" if "longitude" in dem.columns else "lon"
    if elev_col not in dem.columns:
        raise ValueError(f"DEM未找到高程列 {elev_col}；现有列: {list(dem.columns)}")
    dem = dem.rename(columns={lat_dem: "lat", lon_dem: "lon", elev_col: "elev"})
    dem["lat"] = dem["lat"].round(3)
    dem["lon"] = dem["lon"].round(3)
    return dem[["lat","lon","elev"]], elev_min

def list_shapefiles(root):
    pats = [
        os.path.join(root, "*.shp"),
        os.path.join(root, "*.SHP"),
        os.path.join(root, "**", "*.shp"),
        os.path.join(root, "**", "*.SHP"),
    ]
    files = []
    for p in pats:
        files.extend(glob.glob(p, recursive=True))
    # 去重 & 排序
    files = sorted(set(files))
    # 过滤：排除以“边界.shp/边界.SHP”结尾的
    excluded = [f for f in files if os.path.basename(f).endswith("边界.shp") or os.path.basename(f).endswith("边界.SHP")]
    included = [f for f in files if f not in excluded]
    return included, excluded

def to_wgs84(gdf):
    """
    转为 WGS84，经纬度。若缺 CRS：
      - 若坐标绝对值明显大于经纬度范围（>400），尝试设为 EPSG:3413（格陵兰极区投影），再转 4326；
      - 否则直接设为 4326。
    """
    try:
        if gdf.crs is None:
            xmin, ymin, xmax, ymax = gdf.total_bounds
            if max(abs(xmin), abs(ymin), abs(xmax), abs(ymax)) > 400:
                gdf = gdf.set_crs(epsg=3413, allow_override=True)
            else:
                gdf = gdf.set_crs(epsg=4326, allow_override=True)
        return gdf.to_crs(epsg=4326)
    except Exception as e:
        print("CRS处理失败，跳过自动投影；原因：", e)
        return gdf  # 让它照原样画

# ========== 列出 shp ==========
import geopandas as gpd
shp_included, shp_excluded = list_shapefiles(basemap_root)

if len(shp_included) == 0:
    print("⚠ 未找到可用 .shp（排除 *边界.shp 后为空）：", basemap_root)
else:
    print("✓ 用作底图的 .shp：")
    for f in shp_included:
        print("  +", f)

if len(shp_excluded) > 0:
    print("（已排除的 *边界.shp）：")
    for f in shp_excluded:
        print("  -", f)

basemap_layers = []
for shp in shp_included:
    try:
        gdf = gpd.read_file(shp)
        gdf = to_wgs84(gdf)
        basemap_layers.append((os.path.basename(shp), gdf))
    except Exception as e:
        print("✗ 读取失败：", shp, "原因：", e)

# ========== DEM（用于屏蔽海面，可选） ==========
dem_df, elev_min = read_dem_mask(dem_path, elev_col_in_dem, elev_min_for_plot)

# ========== 主流程 ==========
for tag, mk_path in mk_files.items():
    print(f"\n== 处理 {tag} ==")
    df = pd.read_csv(mk_path)
    df.columns = [c.strip() for c in df.columns]
    df = df.rename(columns={"latitude":"lat","longitude":"lon"})
    if "lat" not in df.columns or "lon" not in df.columns:
        raise ValueError(f"{os.path.basename(mk_path)} 未找到经纬度列；现有列: {list(df.columns)}")

    # 对齐坐标精度
    df["lat"] = df["lat"].round(3)
    df["lon"] = df["lon"].round(3)

    # 合并 DEM（只为做 elev≥0 的掩膜；不画等高线）
    merged = df.merge(dem_df, on=["lat","lon"], how="left")

    # 掩膜：只画陆地/冰盖（可按需关闭）
    use = merged[(merged["elev"].isna()) | (merged["elev"] >= elev_min)].copy()

    # 构网格画底图
    LON, LAT, Ztrend = grid_from_points(use, value_col)
    vmin, vmax = auto_symmetric_vmin_vmax(use[value_col].values, 2, 98)

    # ==== 绘图 ====
    fig, ax = plt.subplots(figsize=(8.5, 7), dpi=dpi)

    # 趋势底图
    im = ax.pcolormesh(LON, LAT, Ztrend, shading="nearest", cmap="RdBu_r", vmin=vmin, vmax=vmax, zorder=1)

    # 叠加 shp 底图（多层；已排除 *边界.shp）
    if len(basemap_layers) > 0:
        for i, (name, gdf) in enumerate(basemap_layers):
            color = edge_colors_cycle[i % len(edge_colors_cycle)]
            try:
                gdf.plot(ax=ax, facecolor="none", edgecolor=color, linewidth=edge_linewidth, alpha=edge_alpha, zorder=3)
            except Exception as e:
                print(f"底图绘制失败：{name}: {e}")
    else:
        print("⚠ 没有可用的 basemap_layers，将不叠加底图。")

    # 显著性点（抽稀）
    if sig_col in use.columns:
        sig_pts = use[(use[sig_col]==1) & use[value_col].notna()]
        if not sig_pts.empty:
            sig_pts = thin_sig_points(sig_pts, stride=sig_stride)
            ax.scatter(sig_pts["lon"], sig_pts["lat"], s=5, c="k", alpha=0.25, linewidths=0,
                       label=f"FDR-significant (/{sig_stride} grid)", zorder=4)

    # 轴/色标/标题
    cb = fig.colorbar(im, ax=ax, shrink=0.9, pad=0.02)
    cb.set_label("Albedo trend per decade (Δ/decade)")

    ax.set_title(f"JJA Albedo Trend ({tag}) with Shapefile Basemap (excl. *边界.shp)")
    ax.set_xlabel("Longitude (°)")
    ax.set_ylabel("Latitude (°)")
    ax.set_xlim(np.nanmin(LON), np.nanmax(LON))
    ax.set_ylim(np.nanmin(LAT), np.nanmax(LAT))
    ax.set_aspect("equal", adjustable="box")

    if sig_col in use.columns and 'sig_pts' in locals() and not sig_pts.empty:
        ax.legend(loc="lower left", frameon=True, fontsize=8)

    fig.tight_layout()
    png_path = os.path.join(out_dir, f"MK_JJA_{tag}_trend_with_shapefiles.png")
    fig.savefig(png_path, dpi=dpi)
    plt.close(fig)
    print("已保存图像：", png_path)

print("\n完成 ✅ 输出在：", out_dir)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.linear_model import RidgeCV
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.ensemble import StackingRegressor

# === Step 1: 加载数据 ===
data_path = r"F:\TUB\greenlandProject\ERA5LAND\merged_allfeature.csv"
df = pd.read_csv(data_path)

# === Step 2: 分区定义 ===
region_map = {
    "R1": ['KPC_L', 'KPC_U', 'THU_L', 'THU_L2', 'THU_U', 'RED_L', 'HUM'],
    "R2": ['CEN', 'NEM', 'EGP', 'NAE'],
    "R3": ['TUN', 'ZAC_A', 'ZAC_L', 'ZAC_U', 'FRE'],
    "R4": ['QAS_A', 'QAS_L', 'QAS_M', 'QAS_U', 'TAS_A', 'TAS_L', 'TAS_U', 'MIT'],
    "R5": ['SDM', 'SDL', 'NUK_U', 'NUK_L', 'NUK_N', 'NUK_K'],
    "R6": ['SWC', 'KAN_U', 'KAN_M', 'KAN_L', 'KAN_B', 'KAN_T', 'DY2'],
    "R7": ['LYN_L', 'LYN_T', 'JAR', 'CP1', 'NSE', 'SCO_L', 'SCO_U', 'WEG_L', 'WEG_B'],
    "R8": ['UPE_L', 'UPE_U', 'NAU']
}
# 可选：移除短期站点
short_term_stations = ['KAN_T', 'RED_L' ]
# df = df[~df['station'].isin(short_term_stations)]  # ← 如有需要可启用

# === Step 3: 添加 region 标签 ===
def assign_region(station):
    # 将 R1, R2, R7 合并为 R127
    if station in region_map['R1'] or station in region_map['R2'] or station in region_map['R7']:
        return 'R127'
    for region, stations in region_map.items():
        if station in stations:
            return region
    return 'Other'

df['region'] = df['station'].apply(assign_region)

# === Step 4: 特征和标签 ===
feature_cols = ['longitude', 'latitude', 'altitude_last_valid', 'sp', 'ssrd', 'str',
                'strd', 't2m', 'u10', 'v10', 'snowc', 'sde', 'asn', 'year', 'month']
X_all = df[feature_cols]
y_all = df['albedo']
region_labels = df['region']

# === Step 5: 堆叠模型定义 ===
base_models = [
    ('rf', RandomForestRegressor(n_estimators=100, random_state=42)),
    ('gbr', GradientBoostingRegressor(n_estimators=100, random_state=42))
]
meta_model = RidgeCV()

stacking_model = make_pipeline(
    StandardScaler(),
    StackingRegressor(estimators=base_models, final_estimator=meta_model, n_jobs=-1)
)

# === Step 6: 显示每个 region 的样本数量 ===
print("\n📦 每个 region 的样本数量：")
region_sample_counts = df['region'].value_counts()
print(region_sample_counts)

# === Step 7: 分区留一验证 ===
region_results = {}

# 更新为 R127
for region in ['R127', 'R3', 'R4', 'R5', 'R6', 'R8']:
    print(f"\n🚩 当前留出验证区域: {region}")

    test_mask = region_labels == region
    X_train, X_test = X_all[~test_mask], X_all[test_mask]
    y_train, y_test = y_all[~test_mask], y_all[test_mask]

    print(f"✅ 训练集样本数: {X_train.shape[0]}, 测试集样本数: {X_test.shape[0]}")

    if X_train.shape[0] == 0 or X_test.shape[0] == 0:
        print(f"⚠️ 跳过 {region}，训练集或测试集为空")
        continue

    # 训练与预测
    stacking_model.fit(X_train, y_train)
    y_pred = stacking_model.predict(X_test)

    # 评估指标
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    region_results[region] = {"R2": r2, "RMSE": rmse}
    print(f"📈 R²: {r2:.4f}, RMSE: {rmse:.4f}")

# === Step 8: 汇总结果 ===
print("\n📊 分区留一验证结果：")
for region, res in region_results.items():
    print(f"{region}: R² = {res['R2']:.4f}, RMSE = {res['RMSE']:.4f}")
```

```python
#-------------------------TABPFN----------每个站点分析----------------------------
import pandas as pd
import torch
from tabpfn import TabPFNRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from scipy import stats
import numpy as np

# 加载数据
df = pd.read_csv(r"F:\TUB\greenlandProject\ERA5LAND\merged_allfeature.csv")

# 特征列和目标列
input_features = [
    "longitude", "latitude", "altitude_last_valid", "sp", "ssrd", "str", "strd",
    "t2m", "u10", "v10", "snowc", "sde", "asn", "year", "month"
]
target_column = "albedo"

# 按站点分组
#grouped = df.groupby("station")
grouped = list(df.groupby("station"))[::-1]

# 保存结果
station_results = {}
too_small_stations = []

# 遍历每个站点
for station_name, group in grouped:
    if len(group) < 20:
        too_small_stations.append((station_name, len(group)))
        continue

    # 提取 X, y
    X = group[input_features].values
    y = group[target_column].values

    # 标准化
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # 训练测试划分
    X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

    # 训练 TabPFN
    model = TabPFNRegressor(device='cuda' if torch.cuda.is_available() else 'cpu')
    model.fit(X_train, y_train)

    # 预测
    y_pred = model.predict(X_test)

    # 计算统计量
    rmse = np.sqrt(np.mean((y_pred - y_test) ** 2))
    r = np.corrcoef(y_test, y_pred)[0, 1]
    r2 = r ** 2

    # 置信区间（95%，假设误差正态分布）
    n = len(y_test)
    stderr = np.std(y_pred - y_test, ddof=1) / np.sqrt(n)
    conf_interval = stats.t.interval(0.95, df=n-1, loc=np.mean(y_pred - y_test), scale=stderr)

    station_results[station_name] = {
        "n_samples": len(group),
        "mean_albedo": np.mean(y),
        "rmse": rmse,
        "r": r,
        "r2": r2,
        "conf_int_low": conf_interval[0],
        "conf_int_high": conf_interval[1]
    }

# 打印最终结果
print("\n📊 TabPFN 每站点预测表现（样本数≥20）:")
for station, res in station_results.items():
    print(f"📍 {station} | 样本: {res['n_samples']} | 均值: {res['mean_albedo']:.4f} | "
          f"RMSE: {res['rmse']:.4f} | R: {res['r']:.4f} | R²: {res['r2']:.4f} | "
          f"95%CI: ({res['conf_int_low']:.4f}, {res['conf_int_high']:.4f})")

# 打印小样本站点
if too_small_stations:
    print("\n⚠️ 样本数过少（<20）的站点列表:")
    for station, n in too_small_stations:
        print(f"📍 {station}: 样本数 = {n}")
```

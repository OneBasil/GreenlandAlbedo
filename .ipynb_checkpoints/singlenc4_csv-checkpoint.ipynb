```python
#        将一日的nc4文件转换为csv文件
import netCDF4 as nc
import pandas as pd

# 设置文件路径
input_file = r"F:\Earthdownload\MERRA2_400.tavg1_2d_rad_Nx.20200102.nc4"
output_csv = r"F:\Earthdownload\MERRA2_output.csv"

# 打开 NetCDF 文件
with nc.Dataset(input_file, 'r') as dataset:
    # 提取变量
    lon = dataset.variables['lon'][:]  # 经度
    lat = dataset.variables['lat'][:]  # 纬度
    time = dataset.variables['time'][:]  # 时间
    albedo = dataset.variables['ALBEDO'][:]  # 地表反照率
    swgdn = dataset.variables['SWGDN'][:]  # 短波辐射

# 创建 CSV 数据
rows = []
for i in range(lon.size):
    for j in range(lat.size):
        rows.append({
            'Longitude': lon[i],
            'Latitude': lat[j],
            'Time': time.tolist(),  # 保留原始时间信息
            'ALBEDO': albedo[0, j, i],  # 提取每组反照率
            'SWGDN': swgdn[0, j, i]  # 提取每组短波辐射
        })

# 创建 DataFrame 并保存为 CSV
df = pd.DataFrame(rows)
df.to_csv(output_csv, index=False)

print("CSV 文件生成完成:", output_csv)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# 读取 CSV 文件
csv_file = r"F:\EarthDownload\zzl\ZZL_Project\dataverse_files\dataverse_files\AWS_latest_locations.csv"  # 替换为你的 CSV 文件路径
df = pd.read_csv(csv_file)

# 提取相关变量
stid = df['stid']
latitudes = df['lat']
longitudes = df['lon']

# 创建绘图
fig, ax = plt.subplots(figsize=(10, 8), subplot_kw={'projection': ccrs.PlateCarree()})

# 设置地图
ax.set_extent([-80, -20, 60, 85], crs=ccrs.PlateCarree())  # 设置地图范围：格陵兰岛区域
ax.add_feature(cfeature.BORDERS, linestyle=':', edgecolor='black')  # 添加国界
ax.add_feature(cfeature.LAND, facecolor='lightgray')  # 添加陆地
ax.add_feature(cfeature.LAKES, edgecolor='blue')  # 添加湖泊
ax.add_feature(cfeature.COASTLINE, edgecolor='black')  # 添加海岸线

# 绘制格陵兰岛
# 如果你有格陵兰岛的 shapefile 文件，可以使用 geopandas 加载
# 例如：gdf = gpd.read_file('greenland_shapefile.shp')
# ax.add_geometries(gdf.geometry, crs=ccrs.PlateCarree(), facecolor='green', edgecolor='black', alpha=0.5)

# 添加经纬度网格线（0.625° × 0.5°）
ax.set_xticks(range(-75, -10, 1))  # 经度范围
ax.set_yticks(range(60, 85, 1))  # 纬度范围
ax.grid(True, which='both', color='gray', linestyle='--', linewidth=0.5)
# 绘制站点位置
ax.scatter(longitudes, latitudes, color='red', marker='o', s=3, transform=ccrs.PlateCarree(), label='Stations')

# 显示图例
ax.legend()

# 设置标题
ax.set_title('Stations on Greenland Map')

# 显示地图
plt.show()
# 计算站点数量
num_stations = len(df)

# 输出站点数量
print(f"添加到图中的站点数量是: {num_stations}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from adjustText import adjust_text  # 自动调整标签位置

# 读取 CSV 文件
csv_file = r"F:\TUB\greenlandProject\NEWdataverse_files\AWS_sites_metadata.csv"  # 替换为你的 CSV 文件路径
df = pd.read_csv(csv_file)

# 提取相关变量
stid = df['site_id']
latitudes = df['latitude_installation']
longitudes = df['longitude_installation']

# 创建绘图
fig, ax = plt.subplots(figsize=(10, 8), subplot_kw={'projection': ccrs.PlateCarree()})

# 设置地图
ax.set_extent([-80, -20, 60, 85], crs=ccrs.PlateCarree())  # 设置地图范围：格陵兰岛区域
ax.add_feature(cfeature.BORDERS, linestyle=':', edgecolor='black')  # 添加国界
ax.add_feature(cfeature.LAND, facecolor='lightgray')  # 添加陆地
ax.add_feature(cfeature.LAKES, edgecolor='blue')  # 添加湖泊
ax.add_feature(cfeature.COASTLINE, edgecolor='black')  # 添加海岸线

# 添加经纬度网格线
ax.set_xticks(range(-75, -10, 5))  # 经度
ax.set_yticks(range(60, 85, 5))  # 纬度
ax.grid(True, which='both', color='gray', linestyle='--', linewidth=0.5)

# 绘制站点位置
ax.scatter(longitudes, latitudes, color='red', marker='o', s=10, transform=ccrs.PlateCarree(), label='Stations')

# **自动调整文本标签**
texts = []  # 存储文本对象
for lon, lat, name in zip(longitudes, latitudes, stid):
    text = ax.text(lon, lat, name, transform=ccrs.PlateCarree(), fontsize=8, ha='center', va='bottom', color='black')
    texts.append(text)

# **使用 adjustText 自动调整**
adjust_text(texts, only_move={'points': 'y', 'texts': 'y'}, arrowprops=dict(arrowstyle="-", color='gray', lw=0.5))

# 显示图例
ax.legend()

# 设置标题
ax.set_title('Stations on Greenland Map')

# 保存高清图片
save_path = r"F:\TUB\greenlandProject\NEWdataverse_files\Greenland_Stations_Map.png"
plt.savefig(save_path, dpi=300, bbox_inches='tight')

# 显示地图
# 保存高清图片
plt.savefig("Greenland_Stations_Map.png", dpi=300, bbox_inches='tight')

plt.show()

# 计算站点数量
num_stations = len(df)

# 输出站点数量
print(f"添加到图中的站点数量是: {num_stations}")
```

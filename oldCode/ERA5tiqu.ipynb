```python
#把路径下nc文件都提取albedo值，并按照不同年份保存成csv文件
import xarray as xr
import pandas as pd
import numpy as np

# 文件路径
file_path = r"F:\TUB\greenlandProject\ERA5LAND\forecastAlbedo.nc"
output_folder = r"F:\TUB\greenlandProject\ERA5LAND\forecastalbedo"  # 输出文件夹

# 打开NetCDF文件
ds = xr.open_dataset(file_path, chunks={'time': 1})  # 分块读取

# 确保输出文件夹存在
import os
os.makedirs(output_folder, exist_ok=True)

# 循环处理每年数据
for year in range(1995, 2025):  # 从1995到2024
    print(f"Processing data for year {year}...")

    # 筛选当前年份的数据
    subset_ds = ds.sel(valid_time=slice(f'{year}-01-01', f'{year}-12-31'))

    # 将经纬度保留到小数点后一位
    subset_ds = subset_ds.assign_coords({
        'latitude': np.round(subset_ds['latitude'], 1),
        'longitude': np.round(subset_ds['longitude'], 1)
    })

    # 转换为DataFrame
    df = subset_ds.to_dataframe().reset_index()

    # 仅保留感兴趣的列
    df = df[['valid_time', 'latitude', 'longitude', 'fal']]

    # 删除重复的经纬度组合行
    df = df.drop_duplicates(subset=['valid_time', 'latitude', 'longitude'])

    # 构建输出文件路径
    output_csv_path = os.path.join(output_folder, f"forecastAlbedo_{year}.csv")

    # 保存为CSV文件
    df.to_csv(output_csv_path, index=False)
    print(f"Data for year {year} saved to {output_csv_path}.")

print("All years processed successfully!")
```

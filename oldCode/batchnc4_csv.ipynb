```python
import os
import netCDF4 as nc
import pandas as pd

# 设置文件路径
input_directory = r"F:\EarthDownload"  # 文件夹路径
output_directory = r"F:\EarthDownload"  # 输出文件夹路径，保存转换后的CSV文件

# 获取文件夹下所有的 .nc4 文件
nc4_files = [f for f in os.listdir(input_directory) if f.endswith('.nc4')]

# 遍历所有 .nc4 文件
for nc4_file in nc4_files:
    input_file = os.path.join(input_directory, nc4_file)
    
    # 从文件名中提取日期（假设文件名格式为 MERRA2_401.tavg1_2d_rad_Nx.20200930.nc4）
    date_str = nc4_file.split('.')[-2]  # 获取倒数第二部分，作为日期
    output_csv = os.path.join(output_directory, f"{date_str}.csv")  # 生成对应的 CSV 文件名

    # 打开 NetCDF 文件
    with nc.Dataset(input_file, 'r') as dataset:
        # 提取变量
        lon = dataset.variables['lon'][:]  # 经度
        lat = dataset.variables['lat'][:]  # 纬度
        time = dataset.variables['time'][:]  # 时间
        albedo = dataset.variables['ALBEDO'][:]  # 地表反照率
        swgdn = dataset.variables['SWGDN'][:]  # 短波辐射

    # 创建 CSV 数据
    rows = []
    for i in range(lon.size):
        for j in range(lat.size):
            rows.append({
                'Longitude': lon[i],
                'Latitude': lat[j],
                'Time': time.tolist(),  # 保留原始时间信息
                'ALBEDO': albedo[0, j, i],  # 提取每组反照率
                'SWGDN': swgdn[0, j, i]  # 提取每组短波辐射
            })

    # 创建 DataFrame 并保存为 CSV
    df = pd.DataFrame(rows)
    df.to_csv(output_csv, index=False)

    print(f"CSV 文件生成完成: {output_csv}")
```
